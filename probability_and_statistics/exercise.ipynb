Bài toán:

Cho biến ngẫu nhiên rời rạc $X$ có bảng phân phối xác suất:

| X | 0 | 1 | 2 | 3 |
|---|---:|---:|---:|---:|
| P | a | 0.18 | 0.34 | 2a |

Yêu cầu:

a. Tính trung bình, trung vị, mốt, hệ số biến thiên của $X$  
b. Tính $P(|X-E(X)| \le 1)$ và tìm hàm phân phối xác suất  
c. Thực hiện 3 phép thử độc lập, tính xác suất để có đúng 2 lần $X$ lớn hơn giá trị trung bình của nó

## 1. Biến ngẫu nhiên rời rạc

Biến ngẫu nhiên rời rạc là biến chỉ nhận một số giá trị cụ thể.

Trong bài này:

$$X \in \{0,1,2,3\}$$

Nghĩa là $X$ chỉ có thể nhận các giá trị 0, 1, 2 hoặc 3.

## 2. Bảng phân phối xác suất

Với biến ngẫu nhiên rời rạc, mỗi giá trị của $X$ sẽ đi kèm một xác suất.

Điều kiện bắt buộc:

$$\sum P(X=x_i)=1$$

Tức là tổng tất cả xác suất phải bằng 1.

## 3. Tìm giá trị của $a$

Từ bảng:

$$a + 0.18 + 0.34 + 2a = 1$$

$$3a + 0.52 = 1$$

$$3a = 0.48$$

$$a = 0.16$$

Vậy bảng xác suất đầy đủ là:

| X | 0 | 1 | 2 | 3 |
|---|---:|---:|---:|---:|
| P | 0.16 | 0.18 | 0.34 | 0.32 |

```python

import numpy as np
import pandas as pd

X = np.array([0, 1, 2, 3])
P = np.array([0.16, 0.18, 0.34, 0.32])

df = pd.DataFrame({
    "X": X,
    "P(X)": P,
    "Cumulative P(X)": np.cumsum(P)
})

df
```

## 4. Kỳ vọng / Trung bình

Kỳ vọng còn gọi là **expected value** hoặc **mean**.

Công thức:

$$E(X)=\sum x_iP(x_i)$$

Áp dụng:

$$E(X)=0(0.16)+1(0.18)+2(0.34)+3(0.32)$$

$$E(X)=1.82$$

```python

EX = np.sum(X * P)
EX
```

## 5. Trung vị

Trung vị là giá trị mà xác suất cộng dồn bắt đầu đạt hoặc vượt 0.5.

Bảng cộng dồn:

| X | P(X) | Cộng dồn |
|---|---:|---:|
| 0 | 0.16 | 0.16 |
| 1 | 0.18 | 0.34 |
| 2 | 0.34 | 0.68 |
| 3 | 0.32 | 1.00 |

Vì tại $X=2$, xác suất cộng dồn đạt 0.68, vượt 0.5.

Vậy:

$$\text{Median} = 2$$

```python

cdf = np.cumsum(P)
median = X[np.where(cdf >= 0.5)[0][0]]
median
```

## 6. Mốt

Mốt là giá trị có xác suất lớn nhất.

Trong bảng:

$$P(X=2)=0.34$$

là xác suất lớn nhất.

Vậy:

$$\text{Mode} = 2$$

```python

mode = X[np.argmax(P)]
mode
```

## 7. Phương sai và độ lệch chuẩn

Trước tiên tính:

$$E(X^2)=\sum x_i^2P(x_i)$$

Sau đó:

$$\text{Var}(X)=E(X^2)-[E(X)]^2$$

Độ lệch chuẩn:

$$\sigma = \sqrt{\text{Var}(X)}$$

```python

EX2 = np.sum((X**2) * P)
variance = EX2 - EX**2
std = np.sqrt(variance)

EX2, variance, std
```

Kết quả:

$$E(X^2)=4.42$$

$$\text{Var}(X)=1.1076$$

$$\sigma \approx 1.052$$

## 8. Hệ số biến thiên

Hệ số biến thiên đo mức độ biến động tương đối so với trung bình.

Công thức:

$$CV=\frac{\sigma}{E(X)}$$

Nếu viết theo phần trăm:

$$CV=\frac{\sigma}{E(X)} \times 100\%$$

```python

CV = std / EX
CV_percent = CV * 100

CV, CV_percent
```

Kết quả:

$$CV \approx 0.578$$

hay:

$$CV \approx 57.8\%$$

## 9. Tính $P(|X-E(X)| \le 1)$

Ta có:

$$E(X)=1.82$$

Thay vào:

$$|X-1.82| \le 1$$

Giải bất phương trình:

$$-1 \le X-1.82 \le 1$$

$$0.82 \le X \le 2.82$$

Vì $X$ chỉ nhận các giá trị:

$$0,1,2,3$$

nên các giá trị thỏa điều kiện là:

$$X=1, X=2$$

Do đó:

$$P(|X-E(X)| \le 1)=P(X=1)+P(X=2)$$

$$=0.18+0.34=0.52$$

```python

valid_X = X[(X >= EX - 1) & (X <= EX + 1)]
prob_interval = P[(X >= EX - 1) & (X <= EX + 1)].sum()

valid_X, prob_interval
```

## 10. Hàm phân phối xác suất

Hàm phân phối tích lũy, tiếng Anh là **Cumulative Distribution Function**, viết tắt là CDF.

Định nghĩa:

$$F(x)=P(X \le x)$$

Với bài này:

$$F(x)=\begin{cases}0 & x<0\\0.16 & 0\le x<1\\0.34 & 1\le x<2\\0.68 & 2\le x<3\\1 & x\ge3\end{cases}$$

```python

def F(x):
    return P[X <= x].sum()

test_values = [-1, 0, 0.5, 1, 1.5, 2, 2.5, 3, 4]

cdf_table = pd.DataFrame({
    "x": test_values,
    "F(x) = P(X <= x)": [F(x) for x in test_values]
})

cdf_table
```

## 11. Phân phối nhị thức trong câu c

Câu c nói:

Thực hiện 3 phép thử độc lập, tính xác suất để có đúng 2 lần $X$ lớn hơn giá trị trung bình của nó.

Ta đã có:

$$E(X)=1.82$$

Điều kiện:

$$X > 1.82$$

Các giá trị thỏa:

$$X=2, X=3$$

Xác suất thành công trong một lần thử:

$$p=P(X>1.82)=P(X=2)+P(X=3)$$

$$p=0.34+0.32=0.66$$

Vì có 3 phép thử độc lập và cần đúng 2 lần thành công, ta dùng công thức nhị thức:

$$P(Y=k)=C_n^k p^k(1-p)^{n-k}$$

Trong đó:

- $n=3$
- $k=2$
- $p=0.66$

Áp dụng:

$$P(Y=2)=C_3^2(0.66)^2(1-0.66)$$

$$P(Y=2)=3(0.66)^2(0.34)$$

$$P(Y=2)\approx 0.444$$

```python

from math import comb

n = 3
k = 2
p = P[X > EX].sum()

prob_binomial = comb(n, k) * (p**k) * ((1-p)**(n-k))

p, prob_binomial
```
