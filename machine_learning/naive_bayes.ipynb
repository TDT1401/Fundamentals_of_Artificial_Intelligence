## 1. Naive Bayes là gì?

**Naive Bayes** là một thuật toán **classification** dựa trên xác suất.

Ý tưởng chính:

> Với một dữ liệu mới, ta tính xác suất dữ liệu đó thuộc từng class. Class nào có xác suất cao nhất thì chọn class đó.

Ví dụ bài toán spam email:

- Email thuộc class `spam`?
- Email thuộc class `not spam`?

Model sẽ tính:

$$
P(Spam|Email)
$$

và

$$
P(NotSpam|Email)
$$

Sau đó chọn class có xác suất cao hơn.

## 2. Bayes' Theorem

Công thức Bayes:

$$
P(A|B)=\frac{P(B|A)P(A)}{P(B)}
$$

Trong Naive Bayes, ta thường viết thành:

$$
P(Class|Features)=\frac{P(Features|Class)P(Class)}{P(Features)}
$$

Ý nghĩa:

| Thành phần | Ý nghĩa |
|---|---|
| $P(Class)$ | Xác suất ban đầu của class |
| $P(Features|Class)$ | Xác suất features xuất hiện trong class đó |
| $P(Class|Features)$ | Xác suất cần dự đoán |
| $P(Features)$ | Xác suất features xuất hiện chung |

Trong classification, ta thường chỉ cần so sánh class nào có xác suất cao hơn, nên có thể viết:

$$
P(Class|Features) \propto P(Features|Class)P(Class)
$$

## 3. Naive nghĩa là gì?

**Naive** nghĩa là “ngây thơ”.

Thuật toán này được gọi là naive vì nó giả định rằng các features **độc lập với nhau**.

Đây gọi là:

> **Independence Assumption** = giả định độc lập

Ví dụ email có 3 từ:

```text
free money now
```

Naive Bayes giả định rằng các từ `free`, `money`, `now` độc lập với nhau khi đã biết class là spam:

$$
P(free, money, now|Spam)
= P(free|Spam) \times P(money|Spam) \times P(now|Spam)
$$

Thực tế thì giả định này không hoàn toàn đúng, vì `free` và `money` có thể liên quan với nhau. Nhưng nhờ giả định này, thuật toán chạy nhanh và vẫn rất hiệu quả trong nhiều bài toán text classification.

## 4. Công thức tổng quát

Nếu một dữ liệu có nhiều features:

$$
X = (x_1, x_2, x_3, ..., x_n)
$$

Naive Bayes tính:

$$
P(Class|X) \propto P(Class) \times P(x_1|Class) \times P(x_2|Class) \times ... \times P(x_n|Class)
$$

Nói đơn giản:

> Xác suất class = xác suất ban đầu của class × xác suất từng feature xuất hiện trong class đó.

## 5. Ví dụ tính tay đơn giản

Giả sử có dataset email như sau:

| Email | Có chữ free? | Class |
|---|---|---|
| 1 | Yes | Spam |
| 2 | Yes | Spam |
| 3 | No | Not Spam |
| 4 | Yes | Spam |
| 5 | No | Not Spam |

Ta có:

$$
P(Spam)=\frac{3}{5}
$$

$$
P(NotSpam)=\frac{2}{5}
$$

Với email có chữ `free`:

$$
P(free|Spam)=\frac{3}{3}=1
$$

$$
P(free|NotSpam)=\frac{0}{2}=0
$$

Vậy nếu email có chữ `free`, model có xu hướng dự đoán là **Spam**.

````python
# Demo tính tay xác suất cơ bản

total_emails = 5
spam_count = 3
not_spam_count = 2

p_spam = spam_count / total_emails
p_not_spam = not_spam_count / total_emails

p_free_given_spam = 3 / 3
p_free_given_not_spam = 0 / 2

score_spam = p_spam * p_free_given_spam
score_not_spam = p_not_spam * p_free_given_not_spam

print("P(Spam):", p_spam)
print("P(Not Spam):", p_not_spam)
print("Score Spam:", score_spam)
print("Score Not Spam:", score_not_spam)

if score_spam > score_not_spam:
    print("Prediction: Spam")
else:
    print("Prediction: Not Spam")
````

## 6. Vấn đề xác suất bằng 0 và Laplace Smoothing

Ở ví dụ trên:

$$
P(free|NotSpam)=0
$$

Nếu một feature có xác suất bằng 0, toàn bộ phép nhân có thể bị bằng 0. Đây là vấn đề lớn.

Để xử lý, Naive Bayes thường dùng **Laplace Smoothing**.

Công thức đơn giản:

$$
P(word|Class)=\frac{count(word, Class)+1}{total\ words\ in\ Class + vocabulary\ size}
$$

Ý tưởng:

> Cộng thêm 1 để tránh xác suất bằng 0.

## 7. Các loại Naive Bayes thường gặp

| Loại | Dùng khi nào? |
|---|---|
| Gaussian Naive Bayes | Dữ liệu số liên tục, giả định phân phối chuẩn |
| Multinomial Naive Bayes | Text classification, đếm số lần từ xuất hiện |
| Bernoulli Naive Bayes | Feature dạng 0/1, có hoặc không xuất hiện |

Trong demo email spam, ta dùng **Multinomial Naive Bayes** vì dữ liệu text thường được biểu diễn bằng số lần từ xuất hiện.

## 8. Demo: Spam Classification bằng Naive Bayes

Ta sẽ dùng:

- `CountVectorizer`: chuyển text thành vector số
- `MultinomialNB`: Naive Bayes phù hợp với dữ liệu text

````python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

emails = [
    "free money now",
    "win free prize",
    "free offer available",
    "claim your free reward",
    "hello how are you",
    "meeting schedule today",
    "project discussion tomorrow",
    "please review the report"
]

labels = [
    "spam",
    "spam",
    "spam",
    "spam",
    "not spam",
    "not spam",
    "not spam",
    "not spam"
]

# Chuyển text thành vector đếm số lần từ xuất hiện
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(emails)

# Train model
model = MultinomialNB()
model.fit(X, labels)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())
````

````python
# Test một email mới

test_emails = [
    "free project offer",
    "meeting discussion today",
    "claim free money",
    "please review project report"
]

X_test = vectorizer.transform(test_emails)

predictions = model.predict(X_test)
probabilities = model.predict_proba(X_test)

for email, pred, prob in zip(test_emails, predictions, probabilities):
    print("Email:", email)
    print("Prediction:", pred)
    print("Classes:", model.classes_)
    print("Probabilities:", prob)
    print("-" * 50)
````

## 9. Giải thích kết quả demo

Ví dụ email:

```text
free project offer
```

Email này có cả từ liên quan spam như `free`, `offer`, nhưng cũng có từ bình thường như `project`.

Naive Bayes sẽ tính điểm xác suất cho từng class:

$$
Score(Spam) = P(Spam) \times P(free|Spam) \times P(project|Spam) \times P(offer|Spam)
$$

$$
Score(NotSpam) = P(NotSpam) \times P(free|NotSpam) \times P(project|NotSpam) \times P(offer|NotSpam)
$$

Class nào có score cao hơn thì model chọn class đó.

## 10. So sánh Naive Bayes và KNN

| Tiêu chí | KNN | Naive Bayes |
|---|---|---|
| Ý tưởng | Dựa vào hàng xóm gần nhất | Dựa vào xác suất |
| Cách dự đoán | Xem điểm gần nhất thuộc class nào | Tính xác suất từng class |
| Train | Gần như không train | Có train xác suất |
| Predict | Có thể chậm nếu dữ liệu lớn | Nhanh |
| Giả định | Không có giả định độc lập mạnh | Giả định features độc lập |
| Phù hợp | Dữ liệu số, khoảng cách rõ ràng | Text, spam, sentiment classification |

Tóm lại:

- **KNN** hỏi: dữ liệu này gần nhóm nào nhất?
- **Naive Bayes** hỏi: dữ liệu này có xác suất thuộc nhóm nào cao nhất?

## 11. Demo nhỏ so sánh ý tưởng KNN và Naive Bayes

Giả sử bài toán phân loại email:

- KNN cần biến email thành vector, rồi đo khoảng cách với các email cũ.
- Naive Bayes đếm xác suất các từ xuất hiện trong từng class.

Vì text có rất nhiều từ, Naive Bayes thường rất phù hợp cho bài toán email spam.

````python
from sklearn.neighbors import KNeighborsClassifier

# Dùng lại dữ liệu vector X ở trên
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X, labels)

knn_predictions = knn.predict(X_test)

for email, nb_pred, knn_pred in zip(test_emails, predictions, knn_predictions):
    print("Email:", email)
    print("Naive Bayes Prediction:", nb_pred)
    print("KNN Prediction:", knn_pred)
    print("-" * 50)
````

## 14. Tóm tắt cuối bài

Câu cần nhớ:

> Naive Bayes = Bayes' Theorem + Independence Assumption.

Nó dự đoán bằng cách tính:

$$
P(Class|Features)
$$

Và chọn class có xác suất cao nhất.

Điểm quan trọng nhất:

- `Naive` = giả định các features độc lập
- `Bayes` = dùng Bayes' Theorem
- Rất mạnh cho text classification
- Khác KNN ở chỗ KNN dựa vào distance, còn Naive Bayes dựa vào probability
