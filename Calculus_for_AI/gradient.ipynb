# Gradient và Gradient Descent

Notebook này ghi chú các ý chính về **gradient**, cách hiểu trực giác, công thức cơ bản, liên hệ với Machine Learning và một số demo bằng Python.


## 1. Gradient là gì?

**Gradient** là vector chứa các đạo hàm riêng của một hàm nhiều biến.

Nếu hàm có dạng:

$$f(x, y)$$

thì gradient của hàm là:

$$\nabla f(x, y) = \left[\frac{\partial f}{\partial x}, \frac{\partial f}{\partial y}\right]$$

Trong đó:

- $\frac{\partial f}{\partial x}$: đạo hàm riêng theo biến $x$
- $\frac{\partial f}{\partial y}$: đạo hàm riêng theo biến $y$
- $\nabla f$: ký hiệu gradient


## 2. Ví dụ cơ bản

Cho hàm:

$$f(x, y) = x^2 + y^2$$

Đạo hàm riêng theo $x$:

$$\frac{\partial f}{\partial x} = 2x$$

Đạo hàm riêng theo $y$:

$$\frac{\partial f}{\partial y} = 2y$$

Vậy gradient là:

$$\nabla f(x, y) = [2x, 2y]$$

Tại điểm $(3, 4)$:

$$\nabla f(3, 4) = [6, 8]$$


## 3. Ý nghĩa trực giác của Gradient

Gradient cho biết:

1. Hàm tăng nhanh nhất theo hướng nào.
2. Độ dốc tại điểm đó lớn hay nhỏ.

Hình dung như đang đứng trên một ngọn núi:

- Gradient chỉ hướng leo lên nhanh nhất.
- Đi ngược gradient là hướng đi xuống nhanh nhất.
- Trong Machine Learning, ta thường muốn giảm loss, nên ta đi ngược gradient.


## 4. Gradient khác gì đạo hàm riêng?

Đạo hàm riêng chỉ xét **một biến tại một thời điểm**.

Gradient gom tất cả đạo hàm riêng lại thành **một vector**.

Ví dụ:

$$f(x, y, z) = x^2 + y^2 + z^2$$

Gradient là:

$$\nabla f(x, y, z) = [2x, 2y, 2z]$$

Nói ngắn gọn:

$$\text{Gradient} = \text{vector chứa các đạo hàm riêng}$$


```python
import sympy as sp

# Khai báo biến
x, y = sp.symbols('x y')

# Hàm f(x, y)
f = x**2 + y**2

# Tính đạo hàm riêng
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)

# Gradient
gradient = [df_dx, df_dy]

print('f(x, y) =', f)
print('df/dx =', df_dx)
print('df/dy =', df_dy)
print('Gradient =', gradient)
```

## 5. Tính Gradient tại một điểm bằng Python

Với hàm:

$$f(x, y) = x^2 + y^2$$

Gradient là:

$$\nabla f(x, y) = [2x, 2y]$$

Tại điểm $(3, 4)$:

$$\nabla f(3, 4) = [6, 8]$$


```python
# Tính gradient tại điểm (3, 4)
point = {x: 3, y: 4}

gradient_at_point = [g.subs(point) for g in gradient]

print('Gradient tại (3, 4) =', gradient_at_point)
```

## 6. Gradient Descent là gì?

**Gradient Descent** là thuật toán dùng gradient để tìm điểm làm hàm nhỏ nhất.

Công thức cập nhật cơ bản:

$$x_{new} = x_{old} - \alpha \nabla f(x)$$

Trong đó:

- $x_{old}$: giá trị hiện tại
- $x_{new}$: giá trị sau khi cập nhật
- $\alpha$: learning rate
- $\nabla f(x)$: gradient
- Dấu trừ $-$ nghĩa là đi ngược hướng gradient để giảm giá trị hàm.


## 7. Demo Gradient Descent với hàm một biến

Cho hàm:

$$f(x) = x^2$$

Đạo hàm:

$$f'(x) = 2x$$

Ta muốn tìm điểm nhỏ nhất của hàm. Điểm nhỏ nhất đúng là $x = 0$.

Ta sẽ dùng Gradient Descent để tiến dần về 0.


```python
# Gradient Descent cho f(x) = x^2

def f(x):
    return x**2

def gradient_f(x):
    return 2*x

x_value = 5          # điểm bắt đầu
learning_rate = 0.1  # alpha
steps = 20

history = []

for i in range(steps):
    grad = gradient_f(x_value)
    x_value = x_value - learning_rate * grad
    history.append((i + 1, x_value, f(x_value)))

for step, x_val, loss in history:
    print(f'Step {step:2d}: x = {x_val:.6f}, f(x) = {loss:.6f}')
```

## 8. Vẽ quá trình Gradient Descent


```python
import matplotlib.pyplot as plt

steps_list = [item[0] for item in history]
x_values = [item[1] for item in history]
loss_values = [item[2] for item in history]

plt.figure(figsize=(8, 5))
plt.plot(steps_list, loss_values, marker='o')
plt.xlabel('Step')
plt.ylabel('f(x)')
plt.title('Loss giảm dần qua từng bước Gradient Descent')
plt.grid(True)
plt.show()
```

## 9. Demo Gradient Descent với hàm hai biến

Cho hàm:

$$f(x, y) = x^2 + y^2$$

Gradient:

$$\nabla f(x, y) = [2x, 2y]$$

Công thức cập nhật:

$$x_{new} = x_{old} - \alpha(2x)$$

$$y_{new} = y_{old} - \alpha(2y)$$


```python
def f_2d(x, y):
    return x**2 + y**2

def gradient_2d(x, y):
    return 2*x, 2*y

x_value = 5
y_value = 3
learning_rate = 0.1
steps = 20

history_2d = []

for i in range(steps):
    grad_x, grad_y = gradient_2d(x_value, y_value)
    x_value = x_value - learning_rate * grad_x
    y_value = y_value - learning_rate * grad_y
    loss = f_2d(x_value, y_value)
    history_2d.append((i + 1, x_value, y_value, loss))

for step, x_val, y_val, loss in history_2d:
    print(f'Step {step:2d}: x = {x_val:.6f}, y = {y_val:.6f}, f(x,y) = {loss:.6f}')
```

$$f(x) = (x - 4)^2$$

$$f'(x) = 2(x - 4)$$

## 11. Code mẫu cho bài tập 1


```python
x, y = sp.symbols('x y')

f_ex1 = 3*x**2 + 2*y**2

grad_ex1 = [sp.diff(f_ex1, x), sp.diff(f_ex1, y)]
grad_at_point_ex1 = [g.subs({x: 2, y: 3}) for g in grad_ex1]

print('f(x, y) =', f_ex1)
print('Gradient =', grad_ex1)
print('Gradient tại (2, 3) =', grad_at_point_ex1)
```

## 12. Code mẫu cho bài tập 3


```python
def f_ex3(x):
    return (x - 4)**2

def grad_ex3(x):
    return 2*(x - 4)

x_value = 10
learning_rate = 0.1
steps = 25

for i in range(steps):
    grad = grad_ex3(x_value)
    x_value = x_value - learning_rate * grad
    print(f'Step {i+1:2d}: x = {x_value:.6f}, f(x) = {f_ex3(x_value):.6f}')
```
