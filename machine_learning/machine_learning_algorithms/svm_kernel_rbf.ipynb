# 1. SVM là gì?



**SVM — Support Vector Machine** là một thuật toán Machine Learning dùng phổ biến cho bài toán **classification**.



Ví dụ:



- Email: spam / not spam

- Review: positive / negative

- Bệnh nhân: có bệnh / không có bệnh

- Khách hàng: rủi ro cao / rủi ro thấp



Ý tưởng chính của SVM là:



> Tìm một đường phân chia tốt nhất giữa các nhóm dữ liệu.



Trong không gian 2 chiều, đường phân chia này gọi là:



$$

\text{decision boundary}

$$



Trong không gian nhiều chiều hơn, nó gọi là:



$$

\text{hyperplane}

$$



SVM không chỉ tìm một đường chia được dữ liệu, mà tìm đường chia sao cho **margin lớn nhất**.

# 2. Margin là gì?



**Margin** là khoảng cách từ decision boundary đến các điểm dữ liệu gần nhất của mỗi class.



SVM muốn margin càng rộng càng tốt.



Ý tưởng:



$$

\text{Maximize margin}

$$



Tức là:



> Tạo ra khoảng cách an toàn lớn nhất giữa hai class.



Nếu margin quá nhỏ, model dễ bị ảnh hưởng bởi noise và dễ dự đoán sai trên dữ liệu mới.



Nếu margin rộng, model thường ổn định hơn.

# 3. Support Vectors là gì?

**Support vectors** là những điểm dữ liệu nằm gần decision boundary nhất.

Chúng là các điểm quan trọng nhất vì chính chúng quyết định vị trí của decision boundary.

Nói ngắn gọn:

> Support vectors là những điểm gần ranh giới phân loại nhất và có ảnh hưởng trực tiếp đến cách SVM chia class.

SVM không quan tâm đều đến tất cả điểm dữ liệu. Nó đặc biệt quan tâm đến những điểm nằm sát biên.

# 4. Công thức cơ bản của SVM



Một hyperplane có dạng:



$$

w^T x + b = 0

$$



Trong đó:



- $x$: vector đặc trưng của dữ liệu

- $w$: vector trọng số

- $b$: bias

- $w^T x + b$: hàm quyết định



Khi dự đoán:



$$

f(x) = w^T x + b

$$



Nếu:



$$

f(x) > 0

$$



thì model dự đoán class 1.



Nếu:



$$

f(x) < 0

$$



thì model dự đoán class 0.

# 5. Vì sao cần Kernel Trick?

Không phải dữ liệu nào cũng chia được bằng một đường thẳng.

Ví dụ dữ liệu có dạng vòng tròn:

- Class 1 nằm ở giữa
- Class 0 nằm xung quanh

Trong không gian 2D, rất khó dùng một đường thẳng để chia hai class này.

Đây gọi là dữ liệu **non-linear**, tức là dữ liệu không thể phân tách tuyến tính.

Khi đó, SVM cần một cách để xử lý dữ liệu phi tuyến. Đó là lý do ta dùng **kernel trick**.

# 6. Kernel Trick là gì?



**Kernel trick** là kỹ thuật giúp SVM xử lý dữ liệu phi tuyến bằng cách tính toán như thể dữ liệu đã được đưa lên không gian chiều cao hơn.



Ý tưởng:



$$

(x_1, x_2) \rightarrow (x_1, x_2, x_1^2 + x_2^2)

$$



Trong không gian ban đầu, dữ liệu có thể không chia được bằng đường thẳng.



Nhưng sau khi đưa lên không gian cao hơn, dữ liệu có thể trở nên dễ tách hơn bằng một mặt phẳng.



Điểm hay của kernel trick là:



> Ta không cần thật sự biến đổi dữ liệu sang không gian cao hơn. Kernel chỉ tính mức độ giống nhau giữa hai điểm.



SVM thường cần tính tích vô hướng:



$$

x_i^T x_j

$$



Kernel trick thay tích vô hướng đó bằng hàm kernel:



$$

K(x_i, x_j)

$$



Nói dễ nhớ:



$$

\text{Kernel trick} = \text{Tính như ở không gian cao hơn, nhưng không cần thật sự biến đổi dữ liệu}

$$

# 7. RBF Kernel là gì?



**RBF** là viết tắt của:



$$

\text{Radial Basis Function}

$$



RBF kernel là một kernel rất phổ biến trong SVM, đặc biệt khi dữ liệu phi tuyến.



Công thức:



$$

K(x_i, x_j) = \exp(-\gamma \|x_i - x_j\|^2)

$$



Trong đó:



- $x_i, x_j$: hai điểm dữ liệu

- $\|x_i - x_j\|^2$: khoảng cách bình phương giữa hai điểm

- $\gamma$: tham số điều khiển mức độ ảnh hưởng của mỗi điểm

- $\exp$: hàm mũ, tức là $e^{(\cdot)}$



RBF kernel đo mức độ giống nhau giữa hai điểm dựa trên khoảng cách.



Nếu hai điểm gần nhau:



$$

K(x_i, x_j) \approx 1

$$



Nếu hai điểm xa nhau:



$$

K(x_i, x_j) \approx 0

$$



Tóm tắt:



$$

\text{Gần nhau} \Rightarrow \text{giống nhau nhiều}

$$



$$

\text{Xa nhau} \Rightarrow \text{giống nhau ít}

$$

# 8. Gamma trong RBF Kernel



Tham số quan trọng nhất của RBF kernel là:



$$

\gamma

$$



Gamma kiểm soát **vùng ảnh hưởng** của mỗi điểm dữ liệu.



## Gamma nhỏ



Nếu:



$$

\gamma \text{ nhỏ}

$$



thì mỗi điểm có vùng ảnh hưởng rộng.



Decision boundary thường mượt hơn.



Ưu điểm:



- Ít overfitting hơn

- Model tổng quát hơn



Nhược điểm:



- Nếu quá nhỏ, model có thể quá đơn giản

- Có thể underfitting



## Gamma lớn



Nếu:



$$

\gamma \text{ lớn}

$$



thì mỗi điểm chỉ ảnh hưởng trong vùng rất gần nó.



Decision boundary thường cong và phức tạp hơn.



Ưu điểm:



- Bắt được pattern phức tạp



Nhược điểm:



- Dễ overfitting

- Có thể học thuộc dữ liệu training

# 9. Tham số C trong SVM



Tham số $C$ kiểm soát mức độ phạt khi model phân loại sai.



## C lớn



Nếu:



$$

C \uparrow

$$



model cố gắng phân loại đúng nhiều điểm training nhất có thể.



Kết quả:



- Margin có thể nhỏ hơn

- Boundary có thể phức tạp hơn

- Dễ overfitting hơn



## C nhỏ



Nếu:



$$

C \downarrow

$$



model chấp nhận một số lỗi để có margin rộng hơn.



Kết quả:



- Model mềm hơn

- Ít overfitting hơn

- Nhưng nếu quá nhỏ có thể underfitting



Nói đơn giản:



$$

C \text{ lớn} \Rightarrow \text{model nghiêm khắc hơn}

$$



$$

C \text{ nhỏ} \Rightarrow \text{model linh hoạt hơn}

$$

# 10. So sánh Linear Kernel và RBF Kernel



| Tiêu chí | Linear Kernel | RBF Kernel |
|---|---|---|
| Dạng boundary | Đường thẳng / mặt phẳng | Đường cong phi tuyến |
| Phù hợp với | Dữ liệu gần tuyến tính | Dữ liệu phi tuyến |
| Tốc độ | Nhanh hơn | Chậm hơn |
| Tham số chính | $C$ | $C$, $\gamma$ |
| Nguy cơ overfitting | Thấp hơn | Cao hơn nếu chỉnh sai |
| Khả năng giải thích | Dễ hơn | Khó hơn |



Quy tắc thực tế:



- Nếu dữ liệu gần tuyến tính → thử `kernel="linear"`

- Nếu dữ liệu phi tuyến → thử `kernel="rbf"`

- Luôn cần scale dữ liệu trước khi dùng SVM

# 11. Demo Python: Linear SVM vs RBF SVM

Trong demo này, ta sẽ tạo dữ liệu dạng vòng tròn bằng `make_circles`.

Dữ liệu này không thể chia tốt bằng một đường thẳng.

Ta sẽ so sánh:

1. **Linear SVM**
2. **RBF SVM**

Mục tiêu là quan sát decision boundary của hai model.

````python
# Import thư viện cần thiết

import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
````

````python
# Tạo dữ liệu phi tuyến dạng vòng tròn

X, y = make_circles(
    n_samples=400,
    factor=0.4,
    noise=0.12,
    random_state=42
)

print("Shape of X:", X.shape)
print("Shape of y:", y.shape)
print("Classes:", np.unique(y))
````

````python
# Vẽ dữ liệu ban đầu

plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, edgecolor="k")
plt.title("Dữ liệu dạng vòng tròn - Non-linear data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
````

Nhìn biểu đồ trên, ta thấy dữ liệu có dạng vòng tròn.

Một class nằm ở giữa, class còn lại nằm bên ngoài.

Vì vậy, dùng một đường thẳng để tách dữ liệu sẽ không hiệu quả.

````python
# Chia dữ liệu thành train set và test set

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Train size:", X_train.shape)
print("Test size:", X_test.shape)
````

# 12. Tại sao cần StandardScaler với SVM?



SVM rất nhạy với khoảng cách giữa các điểm dữ liệu.



Nếu một feature có scale lớn hơn feature khác, model có thể bị lệch.



Vì vậy, khi dùng SVM, ta thường dùng:



$$

x' = \frac{x - \mu}{\sigma}

$$



Trong sklearn, ta dùng:



```python

StandardScaler()

```



Để tránh quên bước scale, ta dùng `Pipeline`.

````python
# Tạo Linear SVM

linear_svm = Pipeline([
    ("scaler", StandardScaler()),
    ("model", SVC(kernel="linear", C=1.0))
])

linear_svm.fit(X_train, y_train)

y_pred_linear = linear_svm.predict(X_test)

print("Linear SVM Accuracy:", accuracy_score(y_test, y_pred_linear))
print()
print(classification_report(y_test, y_pred_linear))
````

````python
# Tạo RBF SVM

rbf_svm = Pipeline([
    ("scaler", StandardScaler()),
    ("model", SVC(kernel="rbf", C=1.0, gamma="scale"))
])

rbf_svm.fit(X_train, y_train)

y_pred_rbf = rbf_svm.predict(X_test)

print("RBF SVM Accuracy:", accuracy_score(y_test, y_pred_rbf))
print()
print(classification_report(y_test, y_pred_rbf))
````

# 13. Hàm vẽ Decision Boundary

Ta sẽ viết một hàm để vẽ vùng phân loại của model.

Mục tiêu:

- Vùng màu thể hiện model dự đoán class nào
- Các điểm dữ liệu thật được vẽ lên trên

````python
def plot_decision_boundary(model, X, y, title):
    # Vẽ decision boundary cho mô hình classification có 2 features.

    x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
    y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5

    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 400),
        np.linspace(y_min, y_max, 400)
    )

    grid_points = np.c_[xx.ravel(), yy.ravel()]
    Z = model.predict(grid_points)
    Z = Z.reshape(xx.shape)

    plt.figure(figsize=(7, 5))
    plt.contourf(xx, yy, Z, alpha=0.3)
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolor="k")
    plt.title(title)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.show()
````

````python
# Decision boundary của Linear SVM

plot_decision_boundary(
    linear_svm,
    X,
    y,
    "Linear SVM Decision Boundary"
)
````

````python
# Decision boundary của RBF SVM

plot_decision_boundary(
    rbf_svm,
    X,
    y,
    "RBF SVM Decision Boundary"
)
````

# 16. Ảnh hưởng của Gamma trong RBF Kernel



Bây giờ ta thử các giá trị gamma khác nhau:



- Gamma nhỏ: boundary mượt hơn

- Gamma lớn: boundary phức tạp hơn, dễ overfitting hơn



Ta sẽ thử:



$$

\gamma = 0.1,\ 1,\ 10,\ 100

$$

````python
gammas = [0.1, 1, 10, 100]

for gamma in gammas:
    model = Pipeline([
        ("scaler", StandardScaler()),
        ("model", SVC(kernel="rbf", C=1.0, gamma=gamma))
    ])

    model.fit(X_train, y_train)
    acc = model.score(X_test, y_test)

    plot_decision_boundary(
        model,
        X,
        y,
        f"RBF SVM with gamma={gamma}, Test Accuracy={acc:.3f}"
    )
````

# 17. Ảnh hưởng của C trong SVM



Bây giờ ta giữ RBF kernel và thử các giá trị $C$:



$$

C = 0.01,\ 0.1,\ 1,\ 100

$$



Ý nghĩa:



- $C$ nhỏ: model mềm hơn, chấp nhận lỗi nhiều hơn

- $C$ lớn: model nghiêm khắc hơn, cố gắng phân loại đúng training data hơn

````python
Cs = [0.01, 0.1, 1, 100]

for C in Cs:
    model = Pipeline([
        ("scaler", StandardScaler()),
        ("model", SVC(kernel="rbf", C=C, gamma="scale"))
    ])

    model.fit(X_train, y_train)
    acc = model.score(X_test, y_test)

    plot_decision_boundary(
        model,
        X,
        y,
        f"RBF SVM with C={C}, Test Accuracy={acc:.3f}"
    )
````

# 19. Mini Demo: Tính RBF Kernel giữa hai điểm



Công thức RBF kernel:



$$

K(x_i, x_j) = \exp(-\gamma \|x_i - x_j\|^2)

$$



Ta tự viết hàm RBF kernel để hiểu rõ hơn.

````python
def rbf_kernel(x_i, x_j, gamma):
    # Tính RBF kernel giữa hai điểm.
    distance_squared = np.sum((x_i - x_j) ** 2)
    return np.exp(-gamma * distance_squared)

point_a = np.array([1, 2])
point_b = np.array([1, 3])
point_c = np.array([10, 20])

gamma = 0.1

print("RBF(point_a, point_b):", rbf_kernel(point_a, point_b, gamma))
print("RBF(point_a, point_c):", rbf_kernel(point_a, point_c, gamma))
````

Nhận xét:

- `point_a` và `point_b` gần nhau nên giá trị RBF kernel lớn hơn.
- `point_a` và `point_c` xa nhau nên giá trị RBF kernel nhỏ hơn.

RBF kernel hiểu đơn giản là:

> Hai điểm càng gần nhau thì càng giống nhau. Hai điểm càng xa nhau thì càng ít giống nhau.

# 20. Grid Search để tìm C và Gamma tốt hơn



Trong thực tế, ta thường không chọn $C$ và $\gamma$ bằng cảm tính.



Ta có thể dùng `GridSearchCV` để thử nhiều giá trị và chọn bộ tham số tốt nhất.

````python
from sklearn.model_selection import GridSearchCV

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", SVC(kernel="rbf"))
])

param_grid = {
    "model__C": [0.1, 1, 10, 100],
    "model__gamma": [0.01, 0.1, 1, 10]
}

grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy"
)

grid_search.fit(X_train, y_train)

print("Best parameters:")
print(grid_search.best_params_)

print("\nBest cross-validation accuracy:")
print(grid_search.best_score_)

print("\nTest accuracy:")
print(grid_search.score(X_test, y_test))
````

````python
# Vẽ decision boundary của model tốt nhất

best_model = grid_search.best_estimator_

plot_decision_boundary(
    best_model,
    X,
    y,
    "Best RBF SVM from GridSearchCV"
)
````

# 23. Ghi nhớ nhanh

Nếu phải trả lời ngắn gọn trong bài học hoặc phỏng vấn:

> SVM là thuật toán phân loại bằng cách tìm hyperplane có margin lớn nhất giữa các class.

> Kernel trick giúp SVM xử lý dữ liệu phi tuyến bằng cách tính toán như thể dữ liệu đã được ánh xạ sang không gian chiều cao hơn, nhưng không cần thật sự tính trực tiếp tọa độ mới.

> RBF kernel là kernel đo độ giống nhau giữa hai điểm dựa trên khoảng cách. Hai điểm càng gần nhau thì giá trị kernel càng gần 1, càng xa nhau thì càng gần 0.
