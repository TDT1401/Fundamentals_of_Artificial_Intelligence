```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Đọc dữ liệu
df = pd.read_csv("../data/titanic/Titanic-Dataset.csv")

# Chọn cột dữ liệu thật
data = df["Age"].dropna()

# Tính mean và standard deviation
mean = data.mean()
std = data.std()

print("Mean:", mean)
print("Standard Deviation:", std)

# Vẽ histogram
plt.hist(data, bins=30, density=True, alpha=0.6)

# Tạo trục x
x = np.linspace(data.min(), data.max(), 100)

# Tính Gaussian PDF
y = norm.pdf(x, mean, std)

# Vẽ đường Gaussian
plt.plot(x, y)

plt.xlabel("Age")
plt.ylabel("Density")
plt.title("Gaussian Distribution of Titanic Age")
plt.show()
```
