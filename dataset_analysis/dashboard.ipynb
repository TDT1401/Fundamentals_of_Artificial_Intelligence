# Làm dashboard nhỏ trong Jupyter

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
import os

from IPython.display import display, clear_output
```

```python
# Load dataset Titanic
FILE_NAME = os.path.join("..", "data", "titanic", "Titanic-Dataset.csv")
df = pd.read_csv(FILE_NAME)
print(df.head())
```

```python
# Tạo filter đơn giản

sex_filter = widgets.Dropdown(
    options=["all"] + list(df["Sex"].dropna().unique()),
    value="all",
    description="Gender:"
)

class_filter = widgets.Dropdown(
    options=["all"] + list(df["Pclass"].dropna().unique()),
    value="all",
    description="Class:"
)
```

```python
# Tạo hàm vẽ dashboard

def update_dashboard(sex, passenger_class):
    clear_output(wait=True)
    
    display(sex_filter, class_filter)
    
    data = df.copy()
    
    if sex != "all":
        data = data[data["Sex"] == sex]
        
    if passenger_class != "all":
        data = data[data["Pclass"] == passenger_class]
    
    print("Số dòng dữ liệu:", len(data))
    print("Tỉ lệ sống sót:", round(data["Survived"].mean() * 100, 2), "%")
    print("Tuổi trung bình:", round(data["Age"].mean(), 2))
    print("Giá vé trung bình:", round(data["Fare"].mean(), 2))
    
    plt.figure(figsize=(6, 4))
    sns.countplot(data=data, x="Survived")
    plt.title("Số lượng sống sót và không sống sót")
    plt.show()
    
    plt.figure(figsize=(6, 4))
    sns.histplot(data=data, x="Age", bins=20, kde=True)
    plt.title("Phân bố tuổi")
    plt.show()
    
    plt.figure(figsize=(6, 4))
    sns.boxplot(data=data, x="Pclass", y="Fare")
    plt.title("Giá vé theo hạng vé")
    plt.show()
```

```python
# Kết nối filter với dashboard 
widgets.interactive(
    update_dashboard,
    sex=sex_filter,
    passenger_class=class_filter
)
```
