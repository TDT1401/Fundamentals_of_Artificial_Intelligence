## 1. Gradient Descent là gì?

**Gradient Descent** là thuật toán dùng để tìm giá trị nhỏ nhất của một hàm.

Trong Machine Learning, hàm đó thường là **loss function / cost function**, tức là hàm đo model dự đoán sai nhiều hay ít.

Mục tiêu:

```text
Chỉnh tham số của model sao cho loss nhỏ nhất.
```

Ví dụ:

```text
Loss lớn  -> model dự đoán sai nhiều
Loss nhỏ -> model dự đoán tốt hơn
```



## 2. Trực giác dễ hiểu

Tưởng tượng bạn đang đứng trên một ngọn đồi trong sương mù và muốn đi xuống điểm thấp nhất.

Bạn không nhìn thấy toàn bộ ngọn đồi, nhưng tại vị trí đang đứng, bạn biết hướng nào dốc lên và hướng nào dốc xuống.

**Gradient** cho biết hướng tăng nhanh nhất của hàm.

Vì ta muốn đi xuống điểm thấp nhất, ta phải đi **ngược hướng gradient**.

```text
Gradient        = hướng tăng nhanh nhất
Gradient Descent = đi ngược hướng gradient để giảm loss
```



## 3. Công thức chính

Công thức cập nhật tham số:

$$
\theta_{new} = \theta_{old} - \alpha \nabla J(\theta)
$$

Trong đó:

| Ký hiệu | Ý nghĩa |
|---|---|
| $\theta$ | tham số cần học |
| $J(\theta)$ | loss function |
| $\nabla J(\theta)$ | gradient của loss |
| $\alpha$ | learning rate |

Ý nghĩa:

```text
tham số mới = tham số cũ - learning rate × gradient
```

Dấu trừ rất quan trọng, vì ta muốn đi xuống, không phải đi lên.



## 4. Ví dụ đơn giản với hàm một biến

Cho hàm:

$$
J(x) = x^2
$$

Ta muốn tìm x sao cho $J(x)$ nhỏ nhất.

Dễ thấy:

$$
J(x) = x^2
$$

nhỏ nhất tại:

$$
x = 0
$$

Đạo hàm:

$$
J'(x) = 2x
$$

Công thức cập nhật:

$$
x_{new} = x_{old} - \alpha \times 2x
$$

Nếu bắt đầu từ:

```text
x = 10
learning_rate = 0.1
```

Lần 1:

$$
x_{new} = 10 - 0.1 \times 20 = 8
$$

Lần 2:

$$
x_{new} = 8 - 0.1 \times 16 = 6.4
$$

Ta thấy x đang tiến dần về 0.


````python

# Demo 1: Gradient Descent cho hàm J(x) = x^2

import numpy as np


def loss(x):
    return x**2


def gradient(x):
    return 2 * x


x = 10
learning_rate = 0.1
epochs = 20

history = []

for epoch in range(epochs):
    grad = gradient(x)
    x = x - learning_rate * grad
    current_loss = loss(x)
    history.append((epoch + 1, x, current_loss))
    print(f"Epoch {epoch+1:02d}: x = {x:.6f}, loss = {current_loss:.6f}")
````

````python

# Vẽ quá trình loss giảm dần

import matplotlib.pyplot as plt

loss_values = [item[2] for item in history]

plt.figure(figsize=(8, 5))
plt.plot(range(1, epochs + 1), loss_values, marker='o')
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss giảm dần qua từng epoch")
plt.grid(True)
plt.show()
````


## 5. Learning Rate là gì?

**Learning rate** là độ lớn mỗi bước đi khi cập nhật tham số.

Nếu learning rate quá nhỏ:

```text
Đi rất chậm
Mất nhiều vòng lặp
```

Nếu learning rate quá lớn:

```text
Nhảy quá xa
Có thể vượt qua điểm tối ưu
Loss không giảm hoặc bị tăng
```

Cách hiểu đơn giản:

```text
Learning rate nhỏ quá -> học chậm
Learning rate lớn quá -> học loạn
Learning rate vừa đủ -> học ổn định
```


````python

# Demo 2: So sánh các learning rate khác nhau

learning_rates = [0.001, 0.01, 0.1, 0.9]
initial_x = 10
epochs = 30

plt.figure(figsize=(8, 5))

for lr in learning_rates:
    x = initial_x
    losses = []
    
    for epoch in range(epochs):
        grad = gradient(x)
        x = x - lr * grad
        losses.append(loss(x))
    
    plt.plot(range(1, epochs + 1), losses, marker='o', label=f"lr={lr}")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Ảnh hưởng của learning rate")
plt.legend()
plt.grid(True)
plt.show()
````


## 6. Gradient Descent trong Linear Regression

Với Linear Regression đơn giản:

$$
y_{pred} = wx + b
$$

Model có 2 tham số cần học:

| Tham số | Ý nghĩa |
|---|---|
| $w$ | weight / hệ số góc |
| $b$ | bias / hệ số chặn |

Loss thường dùng là MSE (MSE là Mean Squared Error: sai số bình phương trung bình):

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_{pred} - y)^2
$$

Mục tiêu:

```text
Tìm w và b sao cho MSE nhỏ nhất.
```

Gradient Descent sẽ cập nhật:

$$
w_{new} = w_{old} - \alpha \frac{\partial J}{\partial w}
$$

$$
b_{new} = b_{old} - \alpha \frac{\partial J}{\partial b}
$$


````python

# Demo 3: Linear Regression bằng Gradient Descent

import numpy as np

# Data đơn giản: y = 2x
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# Khởi tạo tham số
w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 1000
n = len(X)

loss_history = []

for epoch in range(epochs):
    y_pred = w * X + b
    loss_value = np.mean((y_pred - y) ** 2)
    loss_history.append(loss_value)
    
    # Gradient theo w và b
    dw = (2/n) * np.sum(X * (y_pred - y))
    db = (2/n) * np.sum(y_pred - y)
    
    # Update tham số
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("w =", w)
print("b =", b)
print("final loss =", loss_history[-1])
````

````python

# Vẽ loss của Linear Regression giảm qua từng epoch

plt.figure(figsize=(8, 5))
plt.plot(loss_history)
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Loss giảm khi train Linear Regression")
plt.grid(True)
plt.show()
````

````python

# So sánh dữ liệu thật và đường dự đoán

plt.figure(figsize=(8, 5))
plt.scatter(X, y, label="Actual data")
plt.plot(X, w * X + b, label="Predicted line")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression bằng Gradient Descent")
plt.legend()
plt.grid(True)
plt.show()
````


## 7. Tóm tắt nhanh

```text
Gradient = hướng tăng nhanh nhất của hàm

Learning Rate = độ dài mỗi bước đi
Epoch = số vòng lặp học
Loss Function = hàm đo model sai bao nhiêu
```

Câu nhớ nhanh:

```text

```

## Bài 1

Cho hàm:

$$
J(x) = x^2 + 4x + 4
$$

Yêu cầu:

1. Tính đạo hàm $J'(x)$
2. Dùng Gradient Descent để tìm x làm $J(x)$ nhỏ nhất
3. Thử với:

```text
x ban đầu = 10
learning_rate = 0.1
epochs = 30
```

Gợi ý:

$$
J(x) = (x + 2)^2
$$

Vậy điểm nhỏ nhất phải gần:

$$
x = -2
$$


````python

# Bài 1 - Code gợi ý

def loss_ex1(x):
    return x**2 + 4*x + 4


def gradient_ex1(x):
    # Đạo hàm của x^2 + 4x + 4 là 2x + 4
    return 2*x + 4


x = 10
learning_rate = 0.1
epochs = 30

for epoch in range(epochs):
    grad = gradient_ex1(x)
    x = x - learning_rate * grad
    print(f"Epoch {epoch+1:02d}: x = {x:.6f}, loss = {loss_ex1(x):.6f}")
````


## Bài 2

Cho data:

```python
X = np.array([1, 2, 3, 4])
y = np.array([3, 6, 9, 12])
```

Yêu cầu:

Dùng Gradient Descent để tìm $w$ và $b$ cho model:

$$
y_{pred} = wX + b
$$

Kết quả kỳ vọng:

```text
w gần 3
b gần 0
```


````python

# Bài 2 - Code gợi ý

X = np.array([1, 2, 3, 4])
y = np.array([3, 6, 9, 12])

w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 1000
n = len(X)

for epoch in range(epochs):
    y_pred = w * X + b
    loss_value = np.mean((y_pred - y) ** 2)
    
    dw = (2/n) * np.sum(X * (y_pred - y))
    db = (2/n) * np.sum(y_pred - y)
    
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("w =", w)
print("b =", b)
print("loss =", loss_value)
````


## Bài 3

Thử thay đổi learning rate:

```text
0.001

1
```

Quan sát:

```text
Loss giảm nhanh hay chậm?

```

Đây là bài rất quan trọng vì trong Machine Learning thật, chọn learning rate sai có thể làm model học rất chậm hoặc không học được.


````python

# Bài 3 - Thử nhiều learning rate cho Linear Regression

X = np.array([1, 2, 3, 4])
y = np.array([3, 6, 9, 12])

learning_rates = [0.001, 0.01, 0.1, 1]
epochs = 50
n = len(X)

plt.figure(figsize=(8, 5))

for lr in learning_rates:
    w = 0.0
    b = 0.0
    losses = []
    
    for epoch in range(epochs):
        y_pred = w * X + b
        loss_value = np.mean((y_pred - y) ** 2)
        losses.append(loss_value)
        
        dw = (2/n) * np.sum(X * (y_pred - y))
        db = (2/n) * np.sum(y_pred - y)
        
        w = w - lr * dw
        b = b - lr * db
    
    plt.plot(losses, label=f"lr={lr}")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("So sánh learning rate trong Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
````
