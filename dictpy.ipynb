# Dictionary (dict)

**Dictionary (dict) là kiểu dữ liệu lưu trữ key–value (cặp khóa–giá trị)**

- Mỗi key là duy nhất (không trùng)
- Key thường là immutable (str, int, tuple…)
- Value có thể là bất kỳ kiểu gì

````python
student = {
    "name": "An",
    "age": 20,
    "major": "IT"
}

print(student["name"])  # An
````

## Các thao tác xử lý dict (cốt lõi phải nắm)

````python
# Thêm / cập nhật phần tử

student["gpa"] = 3.5        # thêm mới
student["age"] = 21         # cập nhật

print(student)  # {'name': 'An', 'age': 21, 'major': 'IT', 'gpa': 3.5}
````

````python
# Xoá phần tử

del student["age"]

student.pop("gpa")          # xóa và trả về giá trị

student.clear()             # xóa toàn bộ
````

````python
# Truy cập an toàn

student.get("name")         # trả về None nếu không có
student.get("score", 0)     # default = 0

print(student.get("name") ) 

# Tránh được lỗi KeyError khi truy cập phần tử không tồn tại
student["score"]  # lỗi nếu không tồn tại
````

````python
# Duyệt dict
for key in student:
    print(key, student[key])

for value in student.values():
    print(value)

for key, value in student.items():
    print(key, value)
````

````python
# Kiểm tra tồn tại

if "name" in student:
    print("Có tồn tại")
````

````python
# lấy danh sách key, value

# student.keys()
student.values()
# student.items()
````

````python
# Copy dict

new_dict = student.copy()

print(new_dict)
````

````python
# Merge dict

a = {"x": 1}
b = {"y": 2}

c = a | b

print(c)  # {'x': 1, 'y': 2}
````

## Các hàm xử lý Dict

**Viết chức năng tìm kiếm trong dict**

````python
# Tìm theo key
def find_by_key(d, key):
    return d.get(key, "Không tìm thấy")

print(find_by_key(student, "name"))  # An
print(find_by_key(student, "score")) # Không tìm thấy
````

````python
# Tìm theo value

def find_by_value(d, target_value):
    result = []
    for key, value in d.items():
        if value == target_value:
            result.append(key)
    return result

print(find_by_value(student, "An"))  # ['name']
print(find_by_value(student, "IT"))  # ['major']
````

````python
# Tìm nâng cao: tìm theo điều kiện

def search_dict(d, condition):
    result = {}
    for key, value in d.items():
        if condition(key, value):
            result[key] = value
    return result

print(search_dict(student, lambda k, v: isinstance(v, int) and v > 20))  # {'age': 21}
````
