## 1. Nhắc lại bản chất đạo hàm

**Đạo hàm** dùng để đo **tốc độ thay đổi** của một hàm số.

Nói dễ hiểu:

- Nếu hàm số thay đổi nhanh → đạo hàm lớn.
- Nếu hàm số thay đổi chậm → đạo hàm nhỏ.
- Nếu hàm số không thay đổi → đạo hàm bằng 0.

Ví dụ:

Nếu:

$$
f(x)=x^2
$$

Thì đạo hàm là:

$$
f'(x)=2x
$$

Ý nghĩa:

- Tại mỗi giá trị của $x$, đạo hàm cho biết đồ thị đang dốc bao nhiêu.
- Trong thực tế, đạo hàm giúp ta biết một đại lượng đang tăng hay giảm nhanh như thế nào.

Ví dụ trong AI:

- Loss function thay đổi theo weight.
- Đạo hàm giúp biết nên tăng hay giảm weight để làm loss nhỏ hơn.

## 2. Công thức gốc của đạo hàm

Công thức định nghĩa chuẩn của đạo hàm là:

$$
f'(x)=\lim_{h\to0}\frac{f(x+h)-f(x)}{h}
$$

Trong đó:

- $f'(x)$: đạo hàm của hàm $f(x)$
- $h$: một thay đổi rất nhỏ của $x$
- $f(x+h)-f(x)$: mức thay đổi của hàm số
- $\frac{f(x+h)-f(x)}{h}$: tốc độ thay đổi trung bình
- Khi $h \to 0$, ta có tốc độ thay đổi tức thời

Tuy nhiên, khi làm bài tập thông thường, ta hiếm khi dùng limit trực tiếp.  
Thay vào đó, ta dùng các **quy tắc đạo hàm** để tính nhanh hơn.

## 3. Các quy tắc tính đạo hàm quan trọng

### 3.1 Đạo hàm của hằng số

Nếu $c$ là một số cố định:

$$
\frac{d}{dx}(c)=0
$$

Ví dụ:

$$
\frac{d}{dx}(5)=0
$$

$$
\frac{d}{dx}(100)=0
$$

Lý do: hằng số không thay đổi theo $x$, nên tốc độ thay đổi bằng 0.

### 3.2 Quy tắc lũy thừa

Đây là công thức rất quan trọng:

$$
\frac{d}{dx}(x^n)=nx^{n-1}
$$

Cách nhớ:

1. Kéo số mũ xuống phía trước.
2. Giảm số mũ đi 1.

Ví dụ:

$$
f(x)=x^3
$$

Kéo số mũ 3 xuống trước, sau đó giảm mũ đi 1:

$$
f'(x)=3x^2
$$

Ví dụ khác:

$$
f(x)=x^5
$$

$$
f'(x)=5x^4
$$

### 3.3 Đạo hàm của $x$

Vì:

$$
x=x^1
$$

Áp dụng quy tắc lũy thừa:

$$
\frac{d}{dx}(x)=1x^{1-1}=1x^0=1
$$

Vậy:

$$
\frac{d}{dx}(x)=1
$$

### 3.4 Đạo hàm của căn

Khi gặp căn, nên đổi về dạng lũy thừa:

$$
\sqrt{x}=x^{1/2}
$$

Áp dụng quy tắc lũy thừa:

$$
\frac{d}{dx}(x^{1/2})=\frac{1}{2}x^{-1/2}
$$

Viết lại đẹp hơn:

$$
\frac{d}{dx}(\sqrt{x})=\frac{1}{2\sqrt{x}}
$$

### 3.5 Đạo hàm của tổng và hiệu

Nếu hàm có nhiều phần cộng hoặc trừ nhau, ta đạo hàm từng phần:

$$
(f(x)+g(x))'=f'(x)+g'(x)
$$

$$
(f(x)-g(x))'=f'(x)-g'(x)
$$

Ví dụ:

$$
f(x)=x^3+2x^2-5x+7
$$

Đạo hàm từng phần:

- $x^3 \rightarrow 3x^2$
- $2x^2 \rightarrow 4x$
- $-5x \rightarrow -5$
- $7 \rightarrow 0$

Kết quả:

$$
f'(x)=3x^2+4x-5
$$

### 3.6 Nhân với hằng số

Nếu một hàm được nhân với một hằng số:

$$
(cf(x))'=cf'(x)
$$

Ví dụ:

$$
f(x)=7x^4
$$

Đạo hàm:

$$
f'(x)=7 \cdot 4x^3=28x^3
$$

### 3.7 Một số đạo hàm cơ bản nên nhớ

| Hàm số | Đạo hàm |
|---|---|
| $e^x$ | $e^x$ |
| $\ln x$ | $\frac{1}{x}$ |
| $\sin x$ | $\cos x$ |
| $\cos x$ | $-\sin x$ |

Các công thức này nên học thuộc dần vì sẽ gặp nhiều trong toán, AI, tối ưu hóa và các mô hình kinh tế.

## 4. Ví dụ step-by-step

### Ví dụ 1

Cho:

$$
f(x)=4x^3-2x^2+7
$$

Đạo hàm từng phần:

$$
4x^3 \rightarrow 12x^2
$$

$$
-2x^2 \rightarrow -4x
$$

$$
7 \rightarrow 0
$$

Kết quả:

$$
f'(x)=12x^2-4x
$$

### Ví dụ 2

Cho:

$$
f(x)=\sqrt{x}+3x^5
$$

Đổi căn sang lũy thừa:

$$
\sqrt{x}=x^{1/2}
$$

Nên:

$$
f(x)=x^{1/2}+3x^5
$$

Đạo hàm từng phần:

$$
x^{1/2} \rightarrow \frac{1}{2}x^{-1/2}
$$

$$
3x^5 \rightarrow 15x^4
$$

Kết quả:

$$
f'(x)=\frac{1}{2\sqrt{x}}+15x^4
$$

### Ví dụ 3

Cho:

$$
f(x)=\frac{1}{x^2}
$$

Đổi về dạng lũy thừa:

$$
\frac{1}{x^2}=x^{-2}
$$

Áp dụng quy tắc lũy thừa:

$$
f'(x)=-2x^{-3}
$$

Viết lại:

$$
f'(x)=\frac{-2}{x^3}
$$

```python
import sympy as sp

# Nếu mục tiêu là tính đạo hàm ký hiệu (symbolic differentiation) thì thường dùng thư viện sympy. 
# Nó cho phép bạn làm việc với các biểu thức toán học một cách trực quan và dễ dàng.
# sympy hiểu toán học dạng biểu thức
```

```python
# Bài 1: f(x)=x4

x = sp.Symbol('x')
f = x**4

df = sp.diff(f, x)
print(df)

"""Giải thích:
 sp.diff(f, x) nghĩa là lấy đạo hàm của hàm f theo biến x.
"""
```

```python
# Bài 2: f(x)=5x3−2x+1

x = sp.Symbol('x')
f = 5*x**3 - 2*x + 1
df = sp.diff(f, x)
print(df)
"""Giải thích:
 Python viết 5x3 là 5*x**3.
"""
```

Bài 3: 
$$
f(x)=\sqrt{x}+x^2

```python
# Bài 3

x = sp.Symbol('x')
f = sp.sqrt(x) + x**2

df = sp.diff(f, x)
print(df)
"""Giải thích:
    sp.sqrt(x) là căn bậc hai của x.
"""
```

Bài 4:
$$
f(x)=\frac{1}{x}
$$

```python
# Bài 4:

x = sp.Symbol('x')
f = 1/x

df = sp.diff(f, x)
print(df)
```

```python
# Bài 5: f(x) = 3*x**5+4*x**2-7
x = sp.Symbol('x')
f = 3*x**5 + 4*x**2 - 7

df = sp.diff(f, x)
print(df)
"""Giải thích: 
    số hằng -7 sau khi đạo hàm sẽ mất vì đạo hàm của hằng số bằng 0.
"""
```
