## 1. Đạo hàm riêng là gì?

Đạo hàm riêng dùng cho **hàm nhiều biến**.

Ví dụ:

$$
f(x,y)=x^2+3xy+y^2
$$

Hàm này phụ thuộc vào cả $x$ và $y$.

Khi lấy đạo hàm riêng:

- Theo $x$: xem $y$ là hằng số
- Theo $y$: xem $x$ là hằng số

Nói đơn giản:

> Đạo hàm riêng cho biết hàm thay đổi như thế nào khi **chỉ một biến thay đổi**, còn các biến khác giữ nguyên.



## 2. Ký hiệu cơ bản

Nếu có hàm:

$$
f(x,y)
$$

Đạo hàm riêng theo $x$:

$$
\frac{\partial f}{\partial x}
$$

Đạo hàm riêng theo $y$:

$$
\frac{\partial f}{\partial y}
$$

Ký hiệu $\partial$ dùng để phân biệt với đạo hàm thường $d$, vì đây là hàm nhiều biến.


## 3. Quy tắc tính nhanh

Muốn tính đạo hàm riêng theo biến nào:

1. Giữ biến đó là biến chính.
2. Các biến còn lại xem như hằng số.
3. Đạo hàm từng hạng tử như đạo hàm bình thường.

Ví dụ:

$$
f(x,y)=x^2+3xy+y^2
$$

Theo $x$:

$$
\frac{\partial f}{\partial x}=2x+3y
$$

Vì:

- $x^2 \rightarrow 2x$
- $3xy \rightarrow 3y$
- $y^2 \rightarrow 0$

Theo $y$:

$$
\frac{\partial f}{\partial y}=3x+2y
$$

Vì:

- $x^2 \rightarrow 0$
- $3xy \rightarrow 3x$
- $y^2 \rightarrow 2y$

## 4. Ví dụ đã giải

Cho:

$$
f(x,y)=3x^2y+2xy^2+y^3
$$

### Đạo hàm riêng theo $x$

Xem $y$ là hằng số:

$$
\frac{\partial f}{\partial x}
=
6xy+2y^2
$$

### Đạo hàm riêng theo $y$

Xem $x$ là hằng số:

$$
\frac{\partial f}{\partial y}
=
3x^2+4xy+3y^2
$$

### Đáp án cuối cùng

$$
\boxed{
\frac{\partial f}{\partial x}=6xy+2y^2
}
$$

$$
\boxed{
\frac{\partial f}{\partial y}=3x^2+4xy+3y^2
}
$$


## 5. Tính đạo hàm riêng bằng Python

Ta dùng thư viện `sympy`.

Trong `sympy`:

```python
sp.diff(f, x)
```

nghĩa là lấy đạo hàm của hàm `f` theo biến `x`.


````python

import sympy as sp

# Khai báo biến
x, y = sp.symbols('x y')

# Khai báo hàm
f = 3*x**2*y + 2*x*y**2 + y**3

# Đạo hàm riêng theo x và y
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)

print("f(x, y) =", f)
print("Đạo hàm riêng theo x =", df_dx)
print("Đạo hàm riêng theo y =", df_dy)
````

## 6. Gradient là gì?

Gradient là vector chứa tất cả đạo hàm riêng.

Nếu:

$$
f(x,y)
$$

thì:

$$
\nabla f =
\left(
\frac{\partial f}{\partial x},
\frac{\partial f}{\partial y}
\right)
$$

Ví dụ với:

$$
f(x,y)=3x^2y+2xy^2+y^3
$$

thì:

$$
\nabla f =
(6xy+2y^2,\ 3x^2+4xy+3y^2)
$$

Trong Machine Learning, gradient rất quan trọng vì nó cho biết hướng để cập nhật tham số trong Gradient Descent.

````python

gradient_f = [sp.diff(f, x), sp.diff(f, y)]

print("Gradient của f là:")
gradient_f
````


### Bài 2


### Bài 3

````python

# Bài 3
f3 = x**2 + 2*x*y + 3*y**2

df3_dx = sp.diff(f3, x)
df3_dy = sp.diff(f3, y)

print("f3 =", f3)
print("df3/dx =", df3_dx)
print("df3/dy =", df3_dy)
````


### Bài 4: Tính đạo hàm riêng tại một điểm

Cho:

$$
f(x,y)=x^2y+3xy^2
$$

Tính:

$$
\frac{\partial f}{\partial x},\quad \frac{\partial f}{\partial y}
$$

tại điểm:

$$
(x,y)=(2,3)
$$


````python

# Bài 4
f4 = x**2*y + 3*x*y**2

df4_dx = sp.diff(f4, x)
df4_dy = sp.diff(f4, y)

# Thay x = 2, y = 3
value_dx = df4_dx.subs({x: 2, y: 3})
value_dy = df4_dy.subs({x: 2, y: 3})

print("f4 =", f4)
print("df4/dx =", df4_dx)
print("df4/dy =", df4_dy)
print("df4/dx tại (2,3) =", value_dx)
print("df4/dy tại (2,3) =", value_dy)
````
