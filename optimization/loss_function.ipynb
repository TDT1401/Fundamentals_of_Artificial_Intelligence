# Loss Function trong Machine Learning

## 1. Loss Function là gì?

**Loss Function** là hàm dùng để đo **model dự đoán sai bao nhiêu so với giá trị thật**.

Nói đơn giản:

- Model dự đoán càng sai → loss càng lớn
- Model dự đoán càng đúng → loss càng nhỏ
- Train model thực chất là quá trình **giảm loss**

Ví dụ:

```text
Giá trị thật = 10
Model dự đoán = 8
Sai số = 10 - 8 = 2
```

Nếu chỉ có một dòng dữ liệu thì dễ tính. Nhưng trong thực tế, dataset có rất nhiều dòng, nên ta cần một công thức để gom toàn bộ lỗi thành **một con số duy nhất**. Con số đó gọi là **loss**.

## 2. Loss Function là khái niệm chung

Các thuật ngữ như:

- MSE
- RMSE
- MAE
- Cross Entropy
- Huber Loss

không chỉ là công thức riêng lẻ. Chúng chính là **các loại loss function khác nhau**.

Hiểu đơn giản:

```text
Loss Function = khái niệm chung

MSE, RMSE, MAE, Cross Entropy = các loại loss function cụ thể
```

Mỗi bài toán khác nhau sẽ dùng loss function khác nhau.

## 3. Bài toán Regression dùng loss nào?

**Regression** là bài toán dự đoán số.

Ví dụ:

- dự đoán giá nhà
- dự đoán doanh thu
- dự đoán tuổi
- dự đoán điểm số
- dự đoán nhiệt độ

Với regression, ta thường dùng:

- MSE
- RMSE
- MAE

## 4. MSE - Mean Squared Error

Công thức:

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2
$$

Trong đó:

- $y_i$: giá trị thật
- $\hat{y}_i$: giá trị model dự đoán
- $n$: số lượng dữ liệu
- $(y_i - \hat{y}_i)$: sai số
- $(y_i - \hat{y}_i)^2$: sai số bình phương

### Vì sao phải bình phương?

Vì nếu chỉ cộng sai số bình thường, sai dương và sai âm có thể triệt tiêu nhau.

Ví dụ:

```text
Sai số: 2, -2, 3
Tổng sai số = 2 + (-2) + 3 = 3
```

Nhưng thực tế model vẫn sai ở cả 3 dòng.

Khi bình phương:

```text
2² = 4
(-2)² = 4
3² = 9
```

Lúc này lỗi không còn bị triệt tiêu.

### Ý nghĩa của MSE

MSE có ý nghĩa là:

```text
Trung bình bình phương sai số của model
```

MSE phạt lỗi lớn rất mạnh.

Ví dụ:

```text
Sai 2  → 2² = 4
Sai 10 → 10² = 100
```

Vì vậy MSE rất nhạy với lỗi lớn và outlier.

````python
import numpy as np

y_true = np.array([10, 20, 30])
y_pred = np.array([8, 22, 27])

errors = y_true - y_pred
squared_errors = errors ** 2
mse = np.mean(squared_errors)

print("Giá trị thật:", y_true)
print("Dự đoán:", y_pred)
print("Sai số:", errors)
print("Sai số bình phương:", squared_errors)
print("MSE:", mse)
````

## 5. MAE - Mean Absolute Error

Công thức:

$$
MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i - \hat{y}_i|
$$

MAE lấy trị tuyệt đối của sai số, sau đó tính trung bình.

Ví dụ:

```text
Sai số: 2, -2, 3
Trị tuyệt đối: 2, 2, 3
MAE = (2 + 2 + 3) / 3 = 2.33
```

### Ý nghĩa của MAE

MAE dễ hiểu hơn MSE vì nó giữ cùng đơn vị với dữ liệu ban đầu.

Ví dụ:

```text
MAE = 2.33
```

Nghĩa là:

```text
Trung bình model dự đoán sai khoảng 2.33 đơn vị
```

Nếu dự đoán giá nhà đơn vị là triệu đồng, thì MAE = 2.33 nghĩa là model sai trung bình khoảng 2.33 triệu đồng.

````python
mae = np.mean(np.abs(y_true - y_pred))

print("Sai số:", errors)
print("Trị tuyệt đối của sai số:", np.abs(errors))
print("MAE:", mae)
````

## 6. RMSE - Root Mean Squared Error

Công thức:

$$
RMSE = \sqrt{MSE}
$$

hoặc viết đầy đủ:

$$
RMSE = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}
$$

RMSE là căn bậc hai của MSE.

### Vì sao cần RMSE?

MSE bị bình phương nên đơn vị cũng bị bình phương.

Ví dụ:

```text
Dữ liệu gốc: triệu đồng
MSE: triệu đồng bình phương
```

Điều này làm MSE khó hiểu với con người.

RMSE lấy căn bậc hai để quay lại đơn vị ban đầu.

### Ý nghĩa của RMSE

Nếu:

```text
RMSE = 5
```

thì có thể hiểu gần đúng:

```text
Model sai khoảng 5 đơn vị
```

RMSE vẫn phạt lỗi lớn mạnh giống MSE, nhưng dễ hiểu hơn MSE.

````python
rmse = np.sqrt(mse)

print("MSE:", mse)
print("RMSE:", rmse)
````

## 7. So sánh MSE, RMSE và MAE

| Loss Function | Dùng cho | Ý nghĩa chính | Đặc điểm |
|---|---|---|---|
| MSE | Regression | Trung bình bình phương sai số | Phạt mạnh lỗi lớn |
| RMSE | Regression | Căn bậc hai của MSE | Dễ hiểu hơn MSE vì cùng đơn vị với dữ liệu |
| MAE | Regression | Sai số tuyệt đối trung bình | Dễ hiểu, ít bị outlier ảnh hưởng hơn MSE |

### Ví dụ về outlier

Giả sử sai số là:

```text
1, 1, 1, 10
```

MAE:

```text
(1 + 1 + 1 + 10) / 4 = 3.25
```

MSE:

```text
(1² + 1² + 1² + 10²) / 4 = 25.75
```

MSE lớn hơn rất nhiều vì nó phạt lỗi 10 cực mạnh.

````python
errors_example = np.array([1, 1, 1, 10])

mae_example = np.mean(np.abs(errors_example))
mse_example = np.mean(errors_example ** 2)
rmse_example = np.sqrt(mse_example)

print("Sai số:", errors_example)
print("MAE:", mae_example)
print("MSE:", mse_example)
print("RMSE:", rmse_example)
````

## 8. Classification dùng loss nào?

**Classification** là bài toán phân loại.

Ví dụ:

- Titanic: sống hoặc không sống
- Email: spam hoặc không spam
- Ảnh: chó hoặc mèo
- Giao dịch: gian lận hoặc không gian lận

Với classification, model thường dự đoán **xác suất**.

Ví dụ Titanic:

```text
Model dự đoán xác suất sống = 0.8
```

Nghĩa là model nghĩ người này có 80% khả năng sống.

Với binary classification, loss phổ biến là **Binary Cross Entropy**.

## 9. Binary Cross Entropy

Công thức:

$$
BCE = -\frac{1}{n}\sum_{i=1}^{n}\left[y_i\log(\hat{y}_i) + (1-y_i)\log(1-\hat{y}_i)\right]
$$

Trong đó:

- $y_i$: nhãn thật, thường là 0 hoặc 1
- $\hat{y}_i$: xác suất model dự đoán
- Nếu $y_i = 1$: người đó thuộc class 1
- Nếu $y_i = 0$: người đó thuộc class 0

### Ý nghĩa

Cross Entropy không đo sai bao nhiêu đơn vị như MSE/MAE.

Nó đo:

```text
Model dự đoán xác suất đúng hay sai?
Model có tự tin sai hay không?
```

Nếu model tự tin đúng → loss thấp.

Nếu model tự tin sai → loss rất cao.

````python
def binary_cross_entropy(y_true, y_pred):
    # Tránh log(0) gây lỗi bằng cách giới hạn xác suất trong khoảng rất nhỏ
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

    return -np.mean(
        y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred)
    )

# Ví dụ 1: Thực tế là 1, model dự đoán xác suất 0.99
y_true_1 = np.array([1])
y_pred_good = np.array([0.99])

# Ví dụ 2: Thực tế là 1, model dự đoán xác suất 0.01
y_pred_bad = np.array([0.01])

loss_good = binary_cross_entropy(y_true_1, y_pred_good)
loss_bad = binary_cross_entropy(y_true_1, y_pred_bad)

print("Thực tế = 1, dự đoán = 0.99, BCE =", loss_good)
print("Thực tế = 1, dự đoán = 0.01, BCE =", loss_bad)
````

## 10. Giải thích kết quả Cross Entropy

Trường hợp 1:

```text
Thực tế = 1
Dự đoán = 0.99
```

Model gần như chắc chắn class đúng là 1. Vì thực tế đúng là 1 nên loss rất thấp.

Trường hợp 2:

```text
Thực tế = 1
Dự đoán = 0.01
```

Model gần như chắc chắn class đúng không phải 1. Nhưng thực tế lại là 1.

Đây là kiểu:

```text
Sai mà còn rất tự tin
```

nên Cross Entropy phạt rất nặng.

## 11. Demo nhỏ: Linear Regression dùng MSE

Trong ví dụ này, ta có dữ liệu:

$$
y = 2x
$$

Ta giả sử model ban đầu là:

$$
\hat{y} = wx
$$

Ban đầu $w$ chưa đúng. Nhiệm vụ của Gradient Descent là chỉnh $w$ để loss giảm dần.

````python
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

w = 0.5
learning_rate = 0.01
epochs = 100

loss_history = []

for epoch in range(epochs):
    y_pred = w * X

    loss = np.mean((y - y_pred) ** 2)
    loss_history.append(loss)

    gradient = np.mean(-2 * X * (y - y_pred))
    w = w - learning_rate * gradient

print("Weight sau khi train:", w)
print("Loss cuối cùng:", loss_history[-1])

plt.plot(loss_history)
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Loss giảm dần qua từng epoch")
plt.show()
````

## 12. Ý nghĩa demo Linear Regression

Ban đầu:

```text
w = 0.5
```

Nhưng dữ liệu thật là:

```text
y = 2x
```

Nên model ban đầu dự đoán sai nhiều.

Sau nhiều epoch, Gradient Descent liên tục chỉnh $w$ để giảm MSE. Kết quả $w$ sẽ tiến gần về 2.

Khi biểu đồ loss giảm dần, điều đó có nghĩa là:

```text
Model đang học tốt hơn qua từng epoch
```

````python
# Bài 1

y_true = np.array([5, 10, 15])
y_pred = np.array([6, 8, 14])

errors = y_true - y_pred
mse_value = np.mean(errors ** 2)
mae_value = np.mean(np.abs(errors))
rmse_value = np.sqrt(mse_value)

print("Sai số:", errors)
print("MSE:", mse_value)
print("MAE:", mae_value)
print("RMSE:", rmse_value)
````

````python
# Bài 2

y_true = np.array([100, 200, 300])
y_pred = np.array([90, 210, 330])

mae_value = np.mean(np.abs(y_true - y_pred))

print("MAE:", mae_value)
print("Ý nghĩa: Model sai trung bình khoảng", mae_value, "đơn vị.")
````

````python
# Bài 3: Hàm tự tính MSE

def mse(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    return np.mean((y_true - y_pred) ** 2)

print(mse([5, 10, 15], [6, 8, 14]))
````

````python
# Bài 4: Hàm tự tính MAE

def mae(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    return np.mean(np.abs(y_true - y_pred))

print(mae([5, 10, 15], [6, 8, 14]))
````
