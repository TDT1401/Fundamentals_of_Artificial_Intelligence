# Python Lists

```
my_list = [1, 2, 3, "hello", 4.5]
```
## Main characteristics:

- Ordered (indexed from 0) - Có thứ tự (index từ 0)
- Mutable → can be modified - sửa được
- Allows duplicate values - Cho phép trùng lặp
- Accessed by index - Truy cập bằng index


````python
my_list = [1, 2, 3, "hello", 4.5]
print(my_list[0])  # 1
````

### 2. Các thao tác xử lý list (cốt lõi cần nhớ)

````python
# Thêm phần tử 

lst = my_list

lst.append(5)        # thêm cuối
lst.insert(1, 10)    # chèn vào vị trí index

print(lst)  # [1, 10, 2, 3, 'hello', 4.5, 5]
````

````python
# Xoá phần tử

lst.remove(10)   # xóa theo giá trị  # [1, 2, 3, 'hello', 4.5, 5]
lst.pop()        # xóa phần tử cuối  # [1, 2, 3, 'hello', 4.5]
lst.pop(1)       # xóa theo index  # [1, 3, 'hello', 4.5]

print(lst)
````

````python
# Duyệt list

for item in lst:
    print(item)
````

````python
# Cắt (slicing)

lst[1:3]   # lấy từ index 1 → 2
lst[:2]    # lấy từ đầu → index 1
````

````python
if 3 in lst:
    print("Có tồn tại")
````

````python
# lst.pop(2)
# print(lst)
# lst.sort()          # tăng dần
lst.sort(reverse=True) # giảm dần
print(lst)
````

### Các hàm xử lý list

````python
# Tính tổng list

def sum_list(lst):
    total = 0
    for num in lst:
        total += num
    return total

print(sum_list(lst))
````

````python
# Tìm phần tử lớn nhất và nhỏ nhất

def find_max_min(lst):
    max_val = lst[0]
    min_val = lst[0]
    
    for num in lst:
        if num > max_val:
            max_val = num
        if num < min_val:
            min_val = num
            
    return max_val, min_val

print(find_max_min(lst))
````

````python
# Lọc các số chẵn

def even_numbers(lst):
    evens = []
    for num in lst:
        if num % 2 == 0:
            evens.append(num)
    return evens

print(even_numbers(lst))
````

````python
# Đếm số lần xuất hiện của phần tử

def count_occurrences(lst, value):
    count = 0
    for item in lst:
        if item == value:
            count += 1
    return count

print(count_occurrences(lst, 3))
````

### Bài tập

**Nhóm cơ bản**
- min_list(lst) → tìm phần tử nhỏ nhất
- average_list(lst) → tính trung bình
- unique_list(lst) → loại bỏ phần tử trùng
- is_sorted(lst) → kiểm tra list đã được sắp xếp chưa
- find_index(lst, x) → tìm vị trí của phần tử

**Nhóm xử lý dữ liệu**
- remove_duplicates(lst) → xóa trùng nhưng giữ thứ tự
- second_largest(lst) → tìm số lớn thứ 2
- merge_lists(lst1, lst2) → gộp 2 list
- intersection(lst1, lst2) → phần tử chung
- difference(lst1, lst2) → phần tử khác nhau

**Nhóm thao tác nâng cao hơn một chút**
- flatten_list(lst) → làm phẳng list lồng nhau.
```
ví dụ: [1, [2,3], 4] → [1,2,3,4]
```

- rotate_list(lst, k) → xoay list k bước.
```
[1,2,3,4], k=1 → [4,1,2,3]
```
- chunk_list(lst, n) → chia list thành các nhóm nhỏ.
```
[1,2,3,4,5], n=2 → [[1,2],[3,4],[5]]
```

- pair_sum(lst, target) → tìm cặp số có tổng = target.
- count_frequency(lst) → đếm số lần xuất hiện của tất cả phần tử.

**Nhóm “tư duy thuật toán”**
- longest_increasing_subsequence(lst)
- max_subarray_sum(lst) (Kadane)
- two_sum(lst, target)
- remove_element(lst, x) (xóa tất cả x)
- binary_search(lst, x) (list đã sort)
- normalize_list(lst) → chuẩn hóa về [0,1]
- standardize_list(lst) → chuẩn hóa z-score
- moving_average(lst, window)
- outlier_detection(lst)
- shuffle_list(lst)
