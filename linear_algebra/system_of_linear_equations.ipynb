# System of linear equations

Công thức chung: AX=b

``` python
import numpy as np

A = np.array([[...], [...]])
b = np.array([...])

x = np.linalg.solve(A, b)
print(x)
```

### Bài 1

$$
\begin{cases}
x + y = 5 \\
x - y = 1
\end{cases}
$$

````python
import numpy as np

A = np.array([
    [1, 1],
    [1, -1]
])

b = np.array([5, 1])

x = np.linalg.solve(A, b)
print(x)
````

### Bài 2
$$

### Bài 4
$$
\begin{cases}
2x - y + z = 3 \\
x + y + z = 6 \\
x - 2y + 3z = 7
\end{cases}
$$

````python
A = np.array([
    [2, -1, 1],
    [1, 1, 1],
    [1, -2, 3]
])

b = np.array([3, 6, 7])

x = np.linalg.solve(A, b)
print(x)
````
