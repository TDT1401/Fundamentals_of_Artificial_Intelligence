```python
# Code thuần Python / Pure Python Code
import math

x = [1, -2, 3]

# L1 norm (cộng giá trị tuyệt đối) / L1 norm (sum of absolute values)
l1 = sum(abs(i) for i in x)

# L2 norm (căn bậc hai của tổng bình phương) / L2 norm (square root of sum of squares)
l2 = math.sqrt(sum(i**2 for i in x))

# L-infinity norm (giá trị tuyệt đối lớn nhất) / L-infinity norm (maximum absolute value)
linf = max(abs(i) for i in x)

print("L1 norm:", l1)
print("L2 norm:", l2)
print("L-infinity norm:", linf)

""" Giải thích / Explanation
- abs(i): Trả về giá trị tuyệt đối của i / Returns the absolute value of i.
- sum(...): Tính tổng của tất cả các phần tử trong iterable / Calculates the sum of all elements in an iterable.
- math.sqrt(...): Tính căn bậc hai của giá trị / Calculates the square root of the value.
- max(...): Trả về giá trị lớn nhất trong iterable / Returns the maximum value in an iterable.
- i**2: Bình phương của i / Square of i.
"""
```

```python
import numpy as np

x = np.array([1, -2, 3])

# L1 norm
l1 = np.linalg.norm(x, ord=1)

# L2 norm (default)
l2 = np.linalg.norm(x)

# L-infinity norm
linf = np.linalg.norm(x, ord=np.inf)

print("L1 norm:", l1)
print("L2 norm:", l2)
print("L-infinity norm:", linf)

""" Giải thích / Explanation
- np.linalg.norm(x) → mặc định là L2 norm / np.linalg.norm(x) → default is L2 norm.
- np.linalg.norm(x, ord=1): Tính L1 norm / Calculates the L1 norm.
- np.linalg.norm(x, ord=np.inf): Tính L-infinity norm / Calculates the L-infinity norm.
"""
```

```python
x = np.array([10, -1, 0.5])

print(np.linalg.norm(x, 1))      # L1
print(np.linalg.norm(x, 2))      # L2
print(np.linalg.norm(x, np.inf)) # L∞
```
