## 1. Bức tranh tổng quan: AI, ML và DL

Quan hệ đúng về mặt học thuật là:

```text
Artificial Intelligence (AI)
└── Machine Learning (ML)
    └── Deep Learning (DL)
```

Nói cách khác:

```text
Deep Learning ⊂ Machine Learning
```

Deep Learning không tách hoàn toàn khỏi Machine Learning.  
Nó là một nhánh con của Machine Learning.

Tuy nhiên, trong thực tế người ta hay tách thành:

```text
Traditional Machine Learning
Deep Learning
```

Cách chia này dùng để phân biệt phương pháp làm việc, không phải để nói rằng Deep Learning nằm ngoài Machine Learning.

## 2. Machine Learning là gì?

Machine Learning là lĩnh vực giúp máy tính học từ dữ liệu thay vì chỉ làm theo luật được lập trình thủ công.

Ví dụ:

- Dữ liệu email → học cách phân loại spam / not spam
- Dữ liệu nhà đất → học cách dự đoán giá nhà
- Dữ liệu khách hàng → học cách dự đoán khách hàng có rời bỏ dịch vụ hay không
- Dữ liệu ảnh → học cách nhận diện mèo, chó, xe, người

Machine Learning thường được chia thành 3 nhóm lớn:

| Nhóm | Ý nghĩa | Ví dụ |
|---|---|---|
| Supervised Learning | Học có đáp án đúng | Classification, Regression |
| Unsupervised Learning | Học không có đáp án đúng | Clustering, PCA |
| Reinforcement Learning | Học bằng thưởng/phạt | Game AI, robot, AlphaGo |

# 3. Supervised Learning

Supervised Learning nghĩa là dữ liệu huấn luyện có cả input `X` và output đúng `y`.

Ví dụ:

```text
Diện tích nhà, số phòng, vị trí → Giá nhà
Nội dung email → Spam / Not Spam
Thông tin bệnh nhân → Sick / Healthy
```

Hai dạng bài toán chính:

| Dạng | Output | Ví dụ |
|---|---|---|
| Regression | Giá trị số | Dự đoán giá nhà, điểm thi, doanh thu |
| Classification | Nhãn / loại | Spam hay không spam, mèo hay chó |

## 3.1 Linear Regression

### Dùng cho

Linear Regression dùng cho bài toán **Regression**, tức là dự đoán giá trị dạng số.

Ví dụ:

- Dự đoán giá nhà
- Dự đoán điểm thi
- Dự đoán doanh thu
- Dự đoán chi phí

### Ý tưởng chính

Linear Regression tìm một đường thẳng hoặc mặt phẳng tốt nhất để biểu diễn mối quan hệ giữa input và output.

Công thức đơn giản:

$$
y = wx + b
$$

Trong đó:

- `y`: giá trị cần dự đoán
- `x`: input
- `w`: weight / slope
- `b`: bias / intercept

### Ưu điểm

- Dễ hiểu
- Dễ học
- Dễ giải thích
- Là nền tảng tốt để học các thuật toán sau

### Nhược điểm

- Không xử lý tốt dữ liệu quá phức tạp
- Giả định mối quan hệ gần tuyến tính giữa input và output

## 3.2 Logistic Regression

### Dùng cho

Logistic Regression dùng cho bài toán **Classification**, đặc biệt là binary classification.

Ví dụ:

- Spam / Not Spam
- Pass / Fail
- Sick / Healthy
- Fraud / Not Fraud

### Ý tưởng chính

Logistic Regression dùng hàm sigmoid để biến kết quả đầu ra thành xác suất.

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

Output của sigmoid nằm trong khoảng:

```text
0 đến 1
```

Ví dụ:

```text
0.8 → khả năng cao thuộc class 1
0.2 → khả năng cao thuộc class 0
```

### Ưu điểm

- Nhanh
- Dễ train
- Dễ giải thích
- Rất phổ biến trong finance, banking, risk và classification cơ bản

### Nhược điểm

- Không mạnh với dữ liệu quá phi tuyến
- Cần feature tốt nếu bài toán phức tạp

## 3.3 Decision Tree

### Dùng cho

Decision Tree dùng cho cả:

- Classification
- Regression

### Ý tưởng chính

Decision Tree chia dữ liệu bằng các câu hỏi điều kiện, giống như cây quyết định.

Ví dụ:

```text
Tuổi < 18?
├── Yes → Student?
│   ├── Yes → Buy
│   └── No → Not Buy
└── No → Income > 50k?
    ├── Yes → Buy
    └── No → Not Buy
```

### Ưu điểm

- Rất dễ hiểu
- Trực quan
- Có thể giải thích được vì sao model đưa ra quyết định
- Dùng tốt với cả dữ liệu số và dữ liệu phân loại

### Nhược điểm

- Dễ overfitting nếu cây quá sâu
- Một thay đổi nhỏ trong dữ liệu có thể làm cây thay đổi nhiều

## 3.4 Random Forest

### Dùng cho

Random Forest dùng cho cả:

- Classification
- Regression

### Ý tưởng chính

Random Forest là tập hợp nhiều Decision Tree.

Thay vì chỉ dùng một cây, model tạo nhiều cây rồi cho các cây “bỏ phiếu”.

Ví dụ:

```text
Tree 1 → Cat
Tree 2 → Dog
Tree 3 → Cat
Tree 4 → Cat

Final prediction → Cat
```

### Ưu điểm

- Chính xác hơn một Decision Tree đơn lẻ
- Giảm overfitting
- Mạnh với nhiều loại dữ liệu
- Rất phổ biến trong Kaggle, finance, medical, business analytics

### Nhược điểm

- Khó giải thích hơn Decision Tree đơn lẻ
- Train chậm hơn Decision Tree

## 3.5 Support Vector Machine (SVM)

### Dùng cho

SVM thường dùng cho Classification, nhưng cũng có biến thể cho Regression.

### Ý tưởng chính

SVM tìm đường phân chia tốt nhất giữa các class.

Trong không gian 2 chiều, đường phân chia có thể viết là:

$$
w^T x + b = 0
$$

SVM cố gắng tạo ra khoảng cách lớn nhất giữa đường phân chia và các điểm dữ liệu gần nhất của mỗi class.

Khoảng cách này gọi là **margin**.

### Ưu điểm

- Mạnh với dataset nhỏ và vừa
- Phân loại tốt nếu dữ liệu có ranh giới rõ
- Có thể xử lý dữ liệu phi tuyến bằng kernel

### Nhược điểm

- Chậm với dataset lớn
- Khó giải thích hơn Logistic Regression
- Cần chọn kernel và tham số phù hợp

## 3.6 K-Nearest Neighbors (KNN)

### Dùng cho

KNN dùng cho:

- Classification
- Regression
- Recommendation cơ bản

### Ý tưởng chính

KNN hoạt động theo logic:

```text
Gần ai thì giống người đó.
```

Ví dụ:

Nếu một điểm dữ liệu mới có 5 hàng xóm gần nhất, trong đó 4 điểm là class “Cat”, thì model dự đoán điểm mới là “Cat”.

### Ưu điểm

- Dễ hiểu
- Không cần training phức tạp
- Trực quan

### Nhược điểm

- Dự đoán chậm nếu dataset lớn
- Nhạy với scale của dữ liệu
- Cần chọn giá trị K phù hợp

## 3.7 Naive Bayes

### Dùng cho

Naive Bayes thường dùng cho:

- Text classification
- Spam detection
- Sentiment analysis
- Document classification

### Ý tưởng chính

Naive Bayes dựa trên định lý Bayes:

$$
P(A|B) = \frac{P(B|A)P(A)}{P(B)}
$$

Trong classification, ta có thể hiểu đơn giản:

```text
Tính xác suất một dữ liệu thuộc từng class
→ Chọn class có xác suất cao nhất
```

### Vì sao gọi là “Naive”?

Vì thuật toán giả định các feature độc lập với nhau.

Ví dụ trong phân loại email spam:

```text
Từ "win", "free", "money" được xem như các dấu hiệu độc lập
```

Giả định này không phải lúc nào cũng đúng, nhưng Naive Bayes vẫn hoạt động rất tốt trong nhiều bài toán text.

### Ưu điểm

- Nhanh
- Đơn giản
- Rất tốt với dữ liệu text
- Cần ít dữ liệu hơn so với nhiều model khác

### Nhược điểm

- Giả định độc lập giữa các feature thường không hoàn toàn đúng
- Không mạnh bằng các model phức tạp trong nhiều bài toán hiện đại

# 4. Unsupervised Learning

Unsupervised Learning nghĩa là dữ liệu không có đáp án đúng `y`.

Model phải tự tìm pattern trong dữ liệu.

Ví dụ:

```text
Dữ liệu khách hàng → tự chia thành các nhóm khách hàng
Dữ liệu sản phẩm → tìm nhóm sản phẩm giống nhau
Dữ liệu nhiều chiều → giảm chiều để dễ phân tích
```

## 4.1 K-Means Clustering

### Dùng cho

K-Means dùng để chia dữ liệu thành K nhóm.

Ví dụ:

- Phân nhóm khách hàng
- Customer segmentation
- Nhóm hành vi người dùng
- Nhóm sản phẩm

### Ý tưởng chính

Quy trình cơ bản:

1. Chọn số nhóm K
2. Random các tâm nhóm gọi là centroid
3. Gán từng điểm dữ liệu vào centroid gần nhất
4. Cập nhật lại centroid
5. Lặp lại cho đến khi ổn định

### Ưu điểm

- Dễ hiểu
- Nhanh
- Phổ biến trong phân tích khách hàng

### Nhược điểm

- Phải chọn K trước
- Nhạy với outlier
- Không tốt nếu nhóm có hình dạng phức tạp

## 4.2 Hierarchical Clustering

### Dùng cho

Hierarchical Clustering dùng để gom nhóm dữ liệu theo dạng cây.

### Ý tưởng chính

Model bắt đầu từ từng điểm riêng lẻ, sau đó dần dần gom các điểm gần nhau thành nhóm lớn hơn.

Ví dụ:

```text
A + B
(A+B) + C
((A+B)+C) + D
```

Kết quả thường được biểu diễn bằng dendrogram.

### Ưu điểm

- Không nhất thiết phải chọn số cụm ngay từ đầu
- Cho thấy cấu trúc phân cấp của dữ liệu

### Nhược điểm

- Chậm với dữ liệu lớn
- Kết quả phụ thuộc vào cách đo khoảng cách

## 4.3 PCA - Principal Component Analysis

### Dùng cho

PCA dùng để giảm số chiều dữ liệu.

Ví dụ:

```text
1000 features → 10 features quan trọng hơn
```

### Ý tưởng chính

PCA tìm các hướng mới trong dữ liệu sao cho giữ lại nhiều thông tin nhất có thể.

### Ứng dụng

- Data compression
- Visualization
- Noise reduction
- Giảm số lượng feature trước khi train model

### Ưu điểm

- Giúp dữ liệu gọn hơn
- Hỗ trợ trực quan hóa dữ liệu nhiều chiều
- Có thể giảm noise

### Nhược điểm

- Khó giải thích ý nghĩa của các principal components
- Có thể mất một phần thông tin

# 5. Deep Learning Algorithms

Deep Learning là một nhánh con của Machine Learning, nhưng thường được tách riêng vì cách hoạt động khác Traditional ML.

Deep Learning dựa trên Neural Network nhiều tầng.

```text
Input
→ Hidden Layer 1
→ Hidden Layer 2
→ Hidden Layer 3
→ Output
```

Từ **Deep** nghĩa là có nhiều tầng xử lý dữ liệu.

## 5.1 Neural Network

### Ý tưởng chính

Neural Network mô phỏng cách các neuron xử lý tín hiệu.

Cấu trúc cơ bản:

```text
Input Layer → Hidden Layers → Output Layer
```

Mỗi layer sẽ học một mức biểu diễn dữ liệu khác nhau.

### Dùng cho

- Classification
- Regression
- Image
- Text
- Audio
- Complex pattern recognition

### Ưu điểm

- Mạnh với dữ liệu phức tạp
- Có thể học pattern mà con người khó tự thiết kế

### Nhược điểm

- Cần nhiều dữ liệu
- Train nặng hơn Traditional ML
- Khó giải thích hơn

## 5.2 CNN - Convolutional Neural Network

### Dùng cho

CNN rất mạnh trong Computer Vision.

Ví dụ:

- Image classification
- Face recognition
- Object detection
- Medical image analysis
- Self-driving car vision

### Ý tưởng chính

CNN học các đặc trưng từ ảnh theo từng tầng:

```text
Cạnh ảnh → Góc cạnh → Hình dạng → Bộ phận → Đối tượng hoàn chỉnh
```

### Ưu điểm

- Rất mạnh với dữ liệu ảnh
- Tự học feature từ ảnh
- Hiệu quả hơn Neural Network thường khi xử lý ảnh

### Nhược điểm

- Cần nhiều dữ liệu ảnh
- Train tốn tài nguyên hơn Traditional ML

## 5.3 RNN / LSTM

### Dùng cho

RNN và LSTM dùng cho dữ liệu dạng chuỗi.

Ví dụ:

- Text
- Speech
- Time series
- Translation
- Stock prediction
- Chatbot đời cũ

### Ý tưởng chính

Dữ liệu chuỗi có thứ tự, ví dụ câu văn:

```text
I love machine learning
```

Từ sau phụ thuộc vào từ trước. RNN/LSTM được thiết kế để xử lý kiểu phụ thuộc theo thời gian này.

### Ưu điểm

- Phù hợp với sequence data
- Có khả năng ghi nhớ thông tin trước đó

### Nhược điểm

- Train khó
- Có thể bị vấn đề vanishing gradient
- Hiện nay nhiều bài toán NLP đã chuyển sang Transformer

## 5.4 Transformer

### Dùng cho

Transformer là nền tảng của nhiều mô hình AI hiện đại.

Ví dụ:

- GPT
- BERT
- Gemini
- Large Language Models
- Chatbot
- Translation
- Summarization
- Generative AI

### Ý tưởng chính

Transformer dùng cơ chế **attention** để xác định phần nào trong dữ liệu là quan trọng.

Trong NLP, attention giúp model hiểu mối quan hệ giữa các từ trong câu tốt hơn.

### Vì sao Transformer quan trọng?

Transformer gần như là nền tảng của làn sóng Generative AI hiện nay.

Các mô hình như ChatGPT hoạt động dựa trên kiến trúc Transformer hoặc biến thể của nó.

### Ưu điểm

- Rất mạnh với text
- Xử lý ngữ cảnh tốt
- Có thể scale lên model rất lớn

### Nhược điểm

- Cần dữ liệu cực lớn
- Cần tài nguyên tính toán lớn
- Khó giải thích chi tiết

# 6. Reinforcement Learning

Reinforcement Learning là nhóm thuật toán học bằng thử sai, thưởng và phạt.

### Ý tưởng chính

```text
Agent thực hiện action
→ Environment phản hồi
→ Agent nhận reward hoặc penalty
→ Agent học cách chọn action tốt hơn
```

### Ví dụ

- AlphaGo
- Game AI
- Robot learning
- Self-driving simulation
- Control system

### Ưu điểm

- Mạnh với bài toán ra quyết định liên tục
- Phù hợp với môi trường có reward rõ ràng

### Nhược điểm

- Khó train
- Cần nhiều thử nghiệm
- Không phải bài toán nào cũng dễ thiết kế reward

# 9. Vì sao nhiều người nghĩ Deep Learning khác Machine Learning?

Lý do là trong thực tế, cách làm Traditional ML và Deep Learning khá khác nhau.

Traditional ML thường cần con người tự thiết kế feature.

Ví dụ phân loại mèo/chó bằng Traditional ML:

```text
Tai nhọn?
Có ria?
Màu lông?
Kích thước?
Hình dạng mặt?
```

Sau đó đưa các feature này vào model:

```text
SVM
Random Forest
Logistic Regression
```

Còn Deep Learning có thể đưa ảnh trực tiếp vào Neural Network.

Model tự học các pattern như:

```text
Cạnh ảnh
Mắt
Tai
Lông
Texture
Hình dạng tổng thể
```

Vì vậy, Deep Learning mạnh hơn Traditional ML trong các bài toán dữ liệu phức tạp như ảnh, âm thanh, video và ngôn ngữ tự nhiên.

# 10. So sánh Traditional ML và Deep Learning

| Traditional Machine Learning | Deep Learning |
|---|---|
| Dùng thuật toán cổ điển | Dùng Neural Network nhiều tầng |
| Thường cần feature engineering | Có thể tự học feature |
| Data vừa và nhỏ vẫn dùng tốt | Thường cần data lớn |
| Train nhanh hơn | Train nặng hơn |
| Dễ explain hơn | Khó explain hơn |
| Mạnh với tabular data | Mạnh với image, text, audio, video |
| Ví dụ: Logistic Regression, SVM, Random Forest | Ví dụ: CNN, RNN, Transformer |

Kết luận:

```text
Deep Learning khác Traditional Machine Learning,
nhưng Deep Learning vẫn thuộc Machine Learning.
```

# 11. Các khái niệm ML vẫn dùng trong Deep Learning

Deep Learning không “thoát ly” khỏi Machine Learning vì nó vẫn dùng các khái niệm nền tảng sau:

- Loss function
- Gradient Descent
- Optimization
- Training
- Epoch
- Batch
- Learning rate
- Overfitting
- Underfitting
- Train / Validation / Test split
- Model evaluation

Ví dụ:

Một Neural Network vẫn cần:

```text
Dữ liệu
→ Model
→ Loss Function
→ Optimizer
→ Training
→ Evaluation
```
