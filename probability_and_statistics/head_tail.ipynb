# Tung đồng xu

```python
import random

coin = random.choice(["Head", "Tail"])

print(coin)
```

```python

for i in range(10):
    result = random.choice(["Head", "Tail"])
    print(result)
```

```python
# đếm số lần xuất hiện của "Head" và "Tail"

heads = 0
tails = 0

for i in range(100):

    result = random.choice(["Head", "Tail"])

    if result == "Head":
        heads += 1
    else:
        tails += 1

print("Heads:", heads)
print("Tails:", tails)
```

```python
# Tính xác suất thực nghiệm

heads = 0
tails = 0
n = 1000

for i in range(n):

    result = random.choice(["Head", "Tail"])

    if result == "Head":
        heads += 1
    else:
        tails += 1

print("Experimental Probability of Head:")
print(heads / n)
```

```python
# Dùng NumPy để mô phỏng việc tung đồng xu nhiều lần và tính xác suất thực nghiệm

import numpy as np

results = np.random.choice(
    ["Head", "Tail"],
    size=1000
)

heads = np.sum(results == "Head")
tails = np.sum(results == "Tail")

print("Heads:", heads)
print("Tails:", tails)

print("P(Head):", heads / 1000)
```

```python
# Visualization using Matplotlib
import matplotlib.pyplot as plt

n = 1000

results = np.random.choice(
    [0, 1],
    size=n
)

running_probability = np.cumsum(results) / np.arange(1, n + 1)

plt.plot(running_probability)

plt.axhline(
    y=0.5,
    linestyle="--"
)

plt.xlabel("Number of Tosses")
plt.ylabel("P(Head)")
plt.title("Coin Toss Simulation")

plt.show()
```
