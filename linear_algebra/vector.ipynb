```python
import numpy as np
```

```python
# tạo vector
A = np.array([1, 2])
B = np.array([3, 4])

print(A)
print(B)
```

```python
# cộng vector 
C = A + B
print(C)
```

```python
# nhân scalar 
D = 3 * A
print(D)
```

```python
# Độ dài vector (magnitude)
length_A = np.linalg.norm(A)
print(length_A)
# đây chính là căn bậc hai của (1^2 + 2^2) = căn bậc hai của 5
```

```python
# Dot product
dot = np.dot(A, B)
print(dot)
```

```python
# Cosine similarity
# độ gióng nhau giữa 2 vector
cosine = np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))
print(cosine)
```

```python
# Ví dụ thực tế
# người A
A = np.array([1, 1, 0])

# người B
B = np.array([1, 1, 0])

# người C
C = np.array([1, 0, 1])

def cosine_sim(x, y):
    return np.dot(x, y) / (np.linalg.norm(x) * np.linalg.norm(y))

print("A vs B:", cosine_sim(A, B))
print("A vs C:", cosine_sim(A, C))
```
