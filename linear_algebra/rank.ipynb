## 1. Rank là gì?

**Rank** của một ma trận là **số lượng hàng hoặc cột độc lập tuyến tính** trong ma trận đó.

Nói dễ hiểu:

> Rank cho biết ma trận có bao nhiêu thông tin thật sự độc lập.

Nếu một hàng hoặc một cột có thể tạo ra từ hàng/cột khác bằng cách nhân hoặc cộng tuyến tính, thì nó **không tạo thêm thông tin mới**.

## 2. Khái niệm độc lập tuyến tính

Một vector được gọi là **phụ thuộc tuyến tính** nếu nó có thể được tạo ra từ vector khác.

Ví dụ:

$$
[2,4] = 2 \times [1,2]
$$

Vậy `[2,4]` không mang thêm thông tin mới so với `[1,2]`.

Ngược lại, nếu không thể tạo vector này từ vector kia, chúng là **độc lập tuyến tính**.

## 3. Hiểu rank bằng trực giác

Có thể hiểu rank là:

> Số hướng độc lập mà ma trận tạo ra.

Ví dụ:

- Rank = 0: không có thông tin, thường là ma trận toàn số 0
- Rank = 1: dữ liệu nằm trên một đường thẳng
- Rank = 2: dữ liệu trải ra một mặt phẳng
- Rank = 3: dữ liệu trải ra không gian 3 chiều

Quy tắc quan trọng:

$$
\text{rank}(A) \leq \min(\text{số hàng}, \text{số cột})
$$

Ví dụ ma trận $3 \times 2$ thì rank tối đa chỉ là 2.

## 4. Cách tính rank bằng tay

Cách phổ biến nhất là dùng **khử Gauss**.

Các bước:

1. Đưa ma trận về dạng bậc thang
2. Đếm số hàng khác 0
3. Số hàng khác 0 chính là rank

Ví dụ:

$$
A =
\begin{bmatrix}
1 & 2 \\
2 & 4
\end{bmatrix}
$$

Ta có:

$$
R_2 = R_2 - 2R_1
$$

$$
\begin{bmatrix}
1 & 2 \\
2 & 4
\end{bmatrix}
\rightarrow
\begin{bmatrix}
1 & 2 \\
0 & 0
\end{bmatrix}
$$

Còn 1 hàng khác 0, nên:

$$
\text{rank}(A) = 1
$$

## 5. Tính rank bằng Python

Ta có thể dùng NumPy để tính rank nhanh:

```python
np.linalg.matrix_rank(A)
```

Hoặc dùng SymPy để xem dạng rút gọn của ma trận.

````python
import numpy as np
import sympy as sp
````

## 6. Ví dụ 1: Rank = 1

$$
A =
\begin{bmatrix}
1 & 2 \\
2 & 4
\end{bmatrix}
$$

Hàng 2 bằng 2 lần hàng 1:

$$
[2,4] = 2 \times [1,2]
$$

Vì vậy chỉ có 1 hàng độc lập, nên rank = 1.

````python
A = np.array([
    [1, 2],
    [2, 4]
])

np.linalg.matrix_rank(A)
````

````python
A_sym = sp.Matrix(A)
A_sym.rref()
````

Kết quả `rref()` cho thấy ma trận sau khi rút gọn chỉ còn 1 hàng có pivot, nên rank = 1.

## 7. Ví dụ 2: Rank = 2

$$
B =
\begin{bmatrix}
1 & 0 \\
0 & 1
\end{bmatrix}
$$

Hai hàng không thể tạo ra từ nhau.

- `[1,0]` là một hướng
- `[0,1]` là một hướng khác

Vậy có 2 hàng độc lập, nên rank = 2.

````python
B = np.array([
    [1, 0],
    [0, 1]
])

np.linalg.matrix_rank(B)
````

## 8. Bài tập bằng Python

Hãy chạy từng ma trận bên dưới và dự đoán rank trước khi xem kết quả.

### Bài 1

````python
M1 = np.array([
    [1, 2],
    [3, 6]
])

print("Rank của M1:", np.linalg.matrix_rank(M1))
sp.Matrix(M1).rref()
````

**Giải thích:**

Hàng 2 = 3 × hàng 1, nên chỉ có 1 hàng độc lập.

$$
\text{rank}(M1) = 1
$$

### Bài 2

````python
M2 = np.array([
    [1, 0],
    [0, 1]
])

print("Rank của M2:", np.linalg.matrix_rank(M2))
sp.Matrix(M2).rref()
````

**Giải thích:**

Hai hàng độc lập với nhau, nên rank = 2.

$$
\text{rank}(M2) = 2
$$

### Bài 3

````python
M3 = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [1, 1, 1]
])

print("Rank của M3:", np.linalg.matrix_rank(M3))
sp.Matrix(M3).rref()
````

**Giải thích:**

Hàng 2 = 2 × hàng 1, nên hàng 2 phụ thuộc.

Hàng 3 không phải là bội của hàng 1, nên nó tạo thêm một thông tin mới.

Vậy có 2 hàng độc lập.

$$
\text{rank}(M3) = 2
$$

### Bài 4

````python
M4 = np.array([
    [1, 2, 3],
    [0, 1, 4],
    [0, 0, 1]
])

print("Rank của M4:", np.linalg.matrix_rank(M4))
sp.Matrix(M4).rref()
````

**Giải thích:**

Ma trận này đã gần như ở dạng bậc thang.

Có 3 hàng khác 0, nên rank = 3.

Vì đây là ma trận 3 × 3 và rank đạt tối đa, nên đây là **full rank matrix**.

$$
\text{rank}(M4) = 3
$$

### Bài 5

````python
M5 = np.array([
    [2, 4, 6],
    [1, 2, 3],
    [3, 6, 9]
])

print("Rank của M5:", np.linalg.matrix_rank(M5))
sp.Matrix(M5).rref()
````

**Giải thích:**

Ta thấy:

$$
[2,4,6] = 2 \times [1,2,3]
$$

$$
[3,6,9] = 3 \times [1,2,3]
$$

Tất cả hàng đều là bội của `[1,2,3]`, nên chỉ có 1 hàng độc lập.

$$
\text{rank}(M5) = 1
$$
