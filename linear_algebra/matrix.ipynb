## 1. Ma trận là gì?

Trong đại số tuyến tính, **ma trận (matrix)** là một bảng số được sắp xếp theo **hàng (row)** và **cột (column)**.

Ví dụ:

$$
A =
\begin{bmatrix}
1 & 2 \\
3 & 4
\end{bmatrix}
$$

Ma trận trên có:

- 2 hàng
- 2 cột
- Kích thước là $2 \times 2$


````python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

print(A)
print("Kích thước của ma trận A:", A.shape)
````

## 2. Các loại ma trận cơ bản

### a) Ma trận hàng

Ma trận hàng chỉ có **1 hàng**.

$$
\begin{bmatrix}
1 & 2 & 3
\end{bmatrix}
$$

Kích thước: $1 \times 3$

````python
row_matrix = np.array([[1, 2, 3]])

print(row_matrix)
print("Kích thước:", row_matrix.shape)
````

### b) Ma trận cột

Ma trận cột chỉ có **1 cột**.

$$
\begin{bmatrix}
1 \\
2 \\
3
\end{bmatrix}
$$

Kích thước: $3 \times 1$

````python
column_matrix = np.array([
    [1],
    [2],
    [3]
])

print(column_matrix)
print("Kích thước:", column_matrix.shape)
````

### c) Ma trận vuông

Ma trận vuông là ma trận có:

$$
\text{số hàng} = \text{số cột}
$$

Ví dụ:

$$
\begin{bmatrix}
1 & 2 \\
3 & 4
\end{bmatrix}
$$

Đây là ma trận vuông cấp 2.

````python
square_matrix = np.array([
    [1, 2],
    [3, 4]
])

print(square_matrix)
print("Kích thước:", square_matrix.shape)
````

### d) Ma trận đơn vị

**Ma trận đơn vị (identity matrix)** là ma trận vuông có:

- Đường chéo chính bằng 1
- Các phần tử còn lại bằng 0

Ví dụ:

$$
I =
\begin{bmatrix}
1 & 0 \\
0 & 1
\end{bmatrix}
$$

Nó giống như số 1 trong phép nhân thông thường.

````python
I = np.eye(2)

print(I)
````

## 3. Các phép tính của ma trận

Ta sử dụng hai ma trận sau để minh họa:

$$
A =
\begin{bmatrix}
1 & 2 \\
3 & 4
\end{bmatrix}
, \quad
B =
\begin{bmatrix}
5 & 6 \\
7 & 8
\end{bmatrix}
$$

````python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("A =")
print(A)

print("\nB =")
print(B)
````

### A. Cộng ma trận

Điều kiện: hai ma trận phải có **cùng kích thước**.

Cộng ma trận là cộng từng phần tử cùng vị trí.

$$
A+B =
\begin{bmatrix}
1+5 & 2+6 \\
3+7 & 4+8
\end{bmatrix}
=
\begin{bmatrix}
6 & 8 \\
10 & 12
\end{bmatrix}
$$

````python
A_plus_B = A + B

print(A_plus_B)
````

### B. Trừ ma trận

Trừ ma trận cũng thực hiện theo từng phần tử cùng vị trí.

$$
A-B =
\begin{bmatrix}
1-5 & 2-6 \\
3-7 & 4-8
\end{bmatrix}
=
\begin{bmatrix}
-4 & -4 \\
-4 & -4
\end{bmatrix}
$$

````python
A_minus_B = A - B

print(A_minus_B)
````

### C. Nhân ma trận với một số

Khi nhân ma trận với một số, ta nhân từng phần tử của ma trận với số đó.

Ví dụ:

$$
2A =
2
\begin{bmatrix}
1 & 2 \\
3 & 4
\end{bmatrix}
=
\begin{bmatrix}
2 & 4 \\
6 & 8
\end{bmatrix}
$$

````python
two_A = 2 * A

print(two_A)
````

### D. Nhân hai ma trận

Đây là phần quan trọng nhất.

Điều kiện để nhân $AB$:

$$
\text{số cột của } A = \text{số hàng của } B
$$

Nếu $A$ có kích thước $m \times n$ và $B$ có kích thước $n \times p$, thì $AB$ có kích thước:

$$
m \times p
$$

Ví dụ:

$$
AB =
\begin{bmatrix}
1\times5 + 2\times7 & 1\times6 + 2\times8 \\
3\times5 + 4\times7 & 3\times6 + 4\times8
\end{bmatrix}
$$

$$
AB =
\begin{bmatrix}
19 & 22 \\
43 & 50
\end{bmatrix}
$$

Logic: lấy **hàng của A** nhân với **cột của B**.

````python
AB = A @ B

print(AB)
````

Trong NumPy, có hai cách phổ biến để nhân ma trận:

```python
A @ B
np.matmul(A, B)
```

Không nên dùng `A * B` nếu mục tiêu là nhân ma trận, vì `A * B` là nhân từng phần tử cùng vị trí.

````python
print("Nhân ma trận bằng A @ B:")
print(A @ B)

print("\nNhân ma trận bằng np.matmul(A, B):")
print(np.matmul(A, B))

print("\nNhân từng phần tử bằng A * B:")
print(A * B)
````

### E. Chuyển vị ma trận

**Transpose** là thao tác đổi hàng thành cột.

Ví dụ:

$$
C =
\begin{bmatrix}
1 & 2 & 3 \\
4 & 5 & 6
\end{bmatrix}
$$

Chuyển vị của $C$ là:

$$
C^T =
\begin{bmatrix}
1 & 4 \\
2 & 5 \\
3 & 6
\end{bmatrix}
$$

````python
C = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("C =")
print(C)

print("\nC chuyển vị =")
print(C.T)
````

### F. Định thức

**Determinant** chỉ áp dụng cho ma trận vuông.

Với ma trận $2 \times 2$:

$$
A =
\begin{bmatrix}
a & b \\
c & d
\end{bmatrix}
$$

Định thức là:

$$
\det(A) = ad - bc
$$

Ví dụ:

$$
A =
\begin{bmatrix}
1 & 2 \\
3 & 4
\end{bmatrix}
$$

$$
\det(A)=1\times4 - 2\times3 = 4 - 6 = -2
$$

Ý nghĩa cơ bản:

- Nếu $\det(A)=0$: ma trận không có nghịch đảo
- Nếu $\det(A)\ne0$: ma trận có nghịch đảo

````python
det_A = np.linalg.det(A)

print(det_A)
````

### G. Ma trận nghịch đảo

Ma trận nghịch đảo giống ý tưởng của phép chia trong số học.

Nếu:

$$
AA^{-1}=I
$$

thì $A^{-1}$ là ma trận nghịch đảo của $A$.

Công thức nghịch đảo cho ma trận $2 \times 2$:

$$
A^{-1}=\frac{1}{ad-bc}
\begin{bmatrix}
d & -b \\
-c & a
\end{bmatrix}
$$

Điều kiện: $ad-bc \ne 0$.

````python
A_inv = np.linalg.inv(A)

print(A_inv)
````

Kiểm tra lại:

$$
AA^{-1}=I
$$

Nếu kết quả gần giống ma trận đơn vị, tức là tính đúng.

````python
check = A @ A_inv

print(check)
````

## 4. Tóm tắt ngắn gọn

| Phép toán | Ý nghĩa |
|---|---|
| Cộng ma trận | Cộng từng phần tử cùng vị trí |
| Trừ ma trận | Trừ từng phần tử cùng vị trí |
| Nhân với số | Nhân từng phần tử với số đó |
| Nhân ma trận | Hàng của ma trận thứ nhất nhân với cột của ma trận thứ hai |
| Chuyển vị | Đổi hàng thành cột |
| Định thức | Kiểm tra ma trận vuông có khả năng nghịch đảo hay không |
| Nghịch đảo | Ma trận đóng vai trò giống phép chia |

Ghi nhớ quan trọng:

- Cộng/trừ: hai ma trận phải cùng kích thước
- Nhân ma trận: số cột của ma trận trước phải bằng số hàng của ma trận sau
- Determinant và inverse chỉ áp dụng cho ma trận vuông
