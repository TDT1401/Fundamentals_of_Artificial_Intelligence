# Numpy & Pandas

## Numpy

```python
import numpy as np

a = np.array([1, 2, 3])
print(a)

# Khác list:
# - list chứa mọi thứ
# - array chỉ chứa 1 kiểu dữ liệu
```

```python
# Array có thể là 1D, 2D, 3D, ... nD

b = np.array([[1, 2, 3],
              [4, 5, 6]])
print(b)
```

```python
# Các thao tác quan trọng
# 1. tạo nhanh
print(np.zeros((2,3)))   # toàn số 0
print(np.ones((2,3)))    # toàn số 1
print(np.arange(0,10))   # giống range
```

```python
# 2. Tính toán
a = np.array([1,2,3])
print(a + 10)     # [11,12,13]
print(a * 2)      # [2,4,6]
# list không làm được điều này

# 3. Truy cập phần tử
print(a[0])       # 1

# 4. thống kê nhanh 
print(np.mean(a))  # .mean() là hàm tính trung bình
print(np.max(a))   # .max() là hàm tính max
print(np.min(a))   # .min() là hàm tính min
print(np.sum(a))   # .sum() là hàm tính tổng
```
