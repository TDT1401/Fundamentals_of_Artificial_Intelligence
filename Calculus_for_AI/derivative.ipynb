# Đạo hàm là gì?

Đạo hàm, tiếng Anh là **derivative**, là một khái niệm rất quan trọng trong toán học, vật lý, kinh tế và đặc biệt là Machine Learning.

Nói đơn giản: **đạo hàm dùng để đo tốc độ thay đổi của một hàm số tại một điểm cụ thể**.

## 1. Hiểu đạo hàm theo trực giác

Giả sử ta có một hàm số:

$$ y = f(x) $$

Trong đó:

- $x$ là đầu vào.
- $y$ là kết quả đầu ra.
- Khi $x$ thay đổi, $y$ cũng có thể thay đổi theo.

Đạo hàm cho ta biết:

> Khi $x$ thay đổi một chút, thì $y$ thay đổi nhanh hay chậm.

Ví dụ dễ hiểu:

- $x$ = thời gian
- $y$ = quãng đường

Khi đó đạo hàm của quãng đường theo thời gian chính là **vận tốc**.

## 2. Định nghĩa toán học của đạo hàm

Công thức định nghĩa đạo hàm là:

$$ f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h} $$

Trong đó:

- $f'(x)$ là đạo hàm của hàm $f(x)$ tại điểm $x$.
- $h$ là một thay đổi rất nhỏ của $x$.
- $f(x+h) - f(x)$ là mức thay đổi của đầu ra $y$.
- $\frac{f(x+h) - f(x)}{h}$ là tốc độ thay đổi trung bình.
- Khi $h \to 0$, ta đang xét tốc độ thay đổi tức thời tại đúng một điểm.

Nói gọn lại:

> Đạo hàm = tốc độ thay đổi tức thời.

## 3. Ý nghĩa hình học của đạo hàm

Trong hình học, đạo hàm tại một điểm chính là **độ dốc của đường tiếp tuyến** tại điểm đó.

Có thể hiểu như sau:

- Nếu đạo hàm $> 0$ thì hàm số đang tăng.
- Nếu đạo hàm $< 0$ thì hàm số đang giảm.
- Nếu đạo hàm $= 0$ thì hàm số có thể đang ở điểm cực đại, cực tiểu hoặc điểm bằng phẳng.

Ví dụ:

- Đường đi lên từ trái sang phải → slope dương → đạo hàm dương.
- Đường đi xuống từ trái sang phải → slope âm → đạo hàm âm.
- Đường nằm ngang → slope bằng 0 → đạo hàm bằng 0.

## 4. Ví dụ đơn giản

Cho hàm số:

$$ f(x) = x^2 $$

Đạo hàm của hàm này là:

$$ f'(x) = 2x $$

Ý nghĩa:

- Tại $x = 1$:

$$ f'(1) = 2 	imes 1 = 2 $$

- Tại $x = 3$:

$$ f'(3) = 2 	imes 3 = 6 $$

Điều này có nghĩa là tại $x = 3$, hàm $f(x) = x^2$ đang tăng nhanh hơn so với tại $x = 1$.

Nói dễ hiểu: càng đi xa khỏi gốc, đường cong $x^2$ càng dốc hơn.

## 5. Một số ký hiệu thường gặp

Các ký hiệu đạo hàm phổ biến:

$$ f'(x) $$

$$ \frac{dy}{dx} $$

$$ \frac{d}{dx} f(x) $$

Ý nghĩa chung của các ký hiệu này đều là:

> Đạo hàm của $y$ theo $x$, hay tốc độ thay đổi của $y$ khi $x$ thay đổi.

## 6. Đạo hàm trong Machine Learning

Trong Machine Learning, đạo hàm cực kỳ quan trọng vì nó giúp model biết phải điều chỉnh tham số theo hướng nào.

Ví dụ, khi training model, ta thường có một hàm lỗi gọi là **loss function**.

Mục tiêu là làm cho loss càng nhỏ càng tốt.

Đạo hàm cho biết:

- Nếu tăng tham số thì loss tăng hay giảm.
- Nên điều chỉnh tham số theo hướng nào.
- Điều chỉnh mạnh hay nhẹ.

Khái niệm liên quan:

> **Gradient** là vector chứa nhiều đạo hàm.

> **Gradient Descent** là thuật toán dùng đạo hàm để tìm điểm làm loss nhỏ nhất.

## 7. Cheat sheet ngắn gọn

| Khái niệm | Ý nghĩa |
|---|---|
| Derivative | Đạo hàm |
| Rate of change | Tốc độ thay đổi |
| Slope | Độ dốc |
| Tangent line | Đường tiếp tuyến |
| $f'(x)$ | Đạo hàm của $f(x)$ |
| $\frac{dy}{dx}$ | Mức thay đổi của $y$ theo $x$ |
| Gradient | Vector gồm nhiều đạo hàm |
| Gradient Descent | Thuật toán tối ưu dùng đạo hàm |

Tóm tắt quan trọng nhất:

> Đạo hàm cho biết một hàm số đang thay đổi nhanh hay chậm tại một điểm.
