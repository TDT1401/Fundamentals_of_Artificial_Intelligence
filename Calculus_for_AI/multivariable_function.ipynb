# Hàm nhiều biến (Multivariable Function)

## 1. Hàm nhiều biến là gì?

Hàm nhiều biến là hàm có **nhiều hơn một biến đầu vào**.

Ví dụ hàm 1 biến:

$$y = f(x)$$

Ví dụ hàm nhiều biến:

$$z = f(x, y)$$

Hoặc tổng quát hơn:

$$f(x_1, x_2, x_3, ..., x_n)$$

Nghĩa là output phụ thuộc vào nhiều input cùng lúc.

## 2. Ví dụ đơn giản

Xét hàm:

$$f(x, y) = x^2 + y^2$$

Hàm này có 2 biến đầu vào là `x` và `y`.

Nếu thay số:

$$f(1,2)=1^2+2^2=5$$

$$f(3,4)=3^2+4^2=25$$

```python
def f(x, y):
    return x**2 + y**2

print(f(1, 2))
print(f(3, 4))
```

## 3. Hàm nhiều biến dùng để làm gì?

Hàm nhiều biến giúp mô tả các bài toán mà kết quả phụ thuộc vào nhiều yếu tố.

Ví dụ:

### Giá nhà

$$Price = f(area, location, age)$$

Giá nhà phụ thuộc vào:

- Diện tích
- Vị trí
- Tuổi căn nhà

### Điểm sinh viên

$$Score = f(math, english, attendance)$$

Điểm số phụ thuộc vào nhiều yếu tố, không chỉ một biến.

## 4. Cách tính với hàm nhiều biến

Bản thân hàm nhiều biến không phải là một phép tính cố định.

Tùy mục tiêu bài toán mà ta dùng công cụ khác nhau:

| Mục tiêu | Công cụ dùng |
|---|---|
| Tính output | Thay số vào hàm |
| Xem từng biến ảnh hưởng thế nào | Đạo hàm riêng |
| Xem hướng tăng nhanh nhất | Gradient |
| Tìm điểm nhỏ nhất / lớn nhất | Đạo hàm riêng + Gradient |
| Tối ưu trong AI | Gradient Descent |

## 5. Đạo hàm riêng trong hàm nhiều biến

Xét hàm:

$$f(x,y)=x^2+3y^2$$

Đạo hàm riêng theo `x`:

$$\frac{\partial f}{\partial x}=2x$$

Đạo hàm riêng theo `y`:

$$\frac{\partial f}{\partial y}=6y$$

Ý nghĩa:

- $\frac{\partial f}{\partial x}$ cho biết nếu chỉ đổi `x` thì hàm thay đổi nhanh thế nào.
- $\frac{\partial f}{\partial y}$ cho biết nếu chỉ đổi `y` thì hàm thay đổi nhanh thế nào.

```python
import sympy as sp

x, y = sp.symbols('x y')

f = x**2 + 3*y**2

df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)

print('f(x,y) =', f)
print('Đạo hàm riêng theo x:', df_dx)
print('Đạo hàm riêng theo y:', df_dy)
```

## 6. Gradient

Gradient là vector gom tất cả đạo hàm riêng lại.

Với hàm:

$$f(x,y)=x^2+3y^2$$

Gradient là:

$$\nabla f(x,y)=\left(\frac{\partial f}{\partial x}, \frac{\partial f}{\partial y}\right)$$

Thay đạo hàm riêng vào:

$$\nabla f(x,y)=(2x, 6y)$$

Ý nghĩa:

- Gradient chỉ hướng hàm tăng nhanh nhất.
- Độ lớn của gradient cho biết hàm tăng mạnh hay yếu.

```python
gradient = [df_dx, df_dy]
print('Gradient:', gradient)

# Tính gradient tại điểm x = 2, y = 1
gradient_at_point = [g.subs({x: 2, y: 1}) for g in gradient]
print('Gradient tại (2,1):', gradient_at_point)
```

## 7. Gradient Descent

Nếu muốn tìm điểm làm hàm nhỏ nhất, ta có thể đi ngược hướng gradient.

Công thức cập nhật:

$$x = x - \eta \frac{\partial f}{\partial x}$$

$$y = y - \eta \frac{\partial f}{\partial y}$$

Trong đó:

- $\eta$ là learning rate
- Gradient chỉ hướng tăng nhanh nhất
- Muốn giảm hàm thì đi ngược hướng gradient

```python
# Demo Gradient Descent cho f(x,y) = x^2 + y^2

def f_value(x, y):
    return x**2 + y**2

def gradient_f(x, y):
    df_dx = 2*x
    df_dy = 2*y
    return df_dx, df_dy

x, y = 5.0, 4.0
learning_rate = 0.1

for i in range(20):
    df_dx, df_dy = gradient_f(x, y)
    x = x - learning_rate * df_dx
    y = y - learning_rate * df_dy
    print(f'Step {i+1}: x={x:.4f}, y={y:.4f}, f(x,y)={f_value(x,y):.4f}')
```

## 8. Liên hệ với AI

Trong AI, model thường có rất nhiều tham số:

$$w_1, w_2, w_3, ..., w_n$$

Loss function thường là hàm nhiều biến:

$$L(w_1, w_2, w_3, ..., w_n)$$

Mục tiêu của AI là tìm bộ tham số làm loss nhỏ nhất.

Quy trình cơ bản:

1. Có loss function
2. Tính đạo hàm riêng theo từng tham số
3. Gom lại thành gradient
4. Dùng gradient descent để update tham số

Nói ngắn gọn:

> Machine Learning và Deep Learning về bản chất là tối ưu một hàm nhiều biến rất lớn.

Yêu cầu:

- Tính $\frac{\partial f}{\partial x}$
- Tính $\frac{\partial f}{\partial y}$
- Viết gradient

### Bài 2

Cho hàm:

$$g(x,y)=3x^2+4xy+y^2$$

Yêu cầu:

- Tính đạo hàm riêng theo `x`
- Tính đạo hàm riêng theo `y`
- Tính gradient tại điểm `(1,2)`

### Bài 3

Dùng Python để chạy Gradient Descent cho hàm:

$$h(x,y)=x^2+2y^2$$

```python
# Gợi ý làm bài tập bằng SymPy

x, y = sp.symbols('x y')

f = x**2 + y**2 + 2*x

print('f(2,3) =', f.subs({x: 2, y: 3}))
print('df/dx =', sp.diff(f, x))
print('df/dy =', sp.diff(f, y))
print('gradient =', [sp.diff(f, x), sp.diff(f, y)])
```
